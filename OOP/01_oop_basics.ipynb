# Introduction to Object Oriented Programming

## Definition

**Object-Oriented Programming (OOP)** is a programming paradigm organized around **objects**—self-contained units that combine **data** (attributes or properties) and **behavior** (methods or functions) that operate on that data.

In OOP, programs are designed by modeling system components as interacting objects, rather than purely as a sequence of steps or functions.

---

## Why It Matters

As software applications grow in size, managing procedural code becomes increasingly difficult. OOP solves this complexity by grouping related variables and functions into single entities.

* **Modularity**: Keeps related data and functions together, making the codebase easier to navigate and maintain.


* **Reusability**: Allows code to be reused across different parts of an application or in future projects through classes and inheritance.


* **Scalability**: Enables large teams to work on separate objects or modules simultaneously without interfering with other parts of the program.


* **Domain Modeling**: Allows developers to construct software architectures that mirror real-world systems and relationships.



---

## Core Building Blocks

### 1. Classes

A **class** is a blueprint or template that defines the data structure and initial state for objects, as well as the behaviors they can perform.

### 2. Objects

An **object** is an instance created from a class blueprint. While the class defines the structure, each object holds its own distinct state (attribute values) while sharing the same methods.

---

## Paradigm Comparison

| Feature | Procedural Programming | Functional Programming | Object-Oriented Programming (OOP) |
| --- | --- | --- | --- |
| **Primary Unit** | Functions / Procedures | Pure Functions | Classes / Objects |
| **Data & Logic Relationship** | Separated (functions operate on standalone data) | Separated (functions transform immutable data) | Bundled together into unified objects |
| **Focus** | How tasks execute step-by-step | What transformations occur on data | How entities interact with each other |
| **State Handling** | Global / Local state altered imperatively | Avoids state changes and side effects | Encapsulated within object instances |

---

## Common Mistakes

* **Creating "God Classes"**: Designing massive classes that handle too many unrelated responsibilities instead of breaking them into smaller, modular objects.
* **Applying OOP Unnecessarily**: Forcing object-oriented structures onto small, linear scripts where simple functions or procedural logic would be cleaner and easier to read.

---

## Quick Recap

* **OOP** structures programs around objects that group **data** (attributes) and **behavior** (methods).


* A **class** is the blueprint; an **object** is the concrete instance created from that blueprint.
* Use OOP when building large, scalable applications or modeling complex domain relationships.

```python
# Creating our first class in Python

class Person: 
    name = "John Doe" # attributes with a default value
    age = 30
    job = "Software Engineer"
    def introduce(self): # method that introduces the person
        print(f"Hello, my name is {self.name}. I am {self.age} years old and I work as a {self.job}.")

person1 = Person() 
person1.name = "Saad Sufyan"
person1.age = 15
person1.job = "Data Analyst"
person1.introduce()
person2 = Person()
person2.introduce()
```

```python
# Creating a Car Class
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def start(self):
        print(f"{self.brand} {self.model} has started.")


car = Car("Toyota", "Corolla")

print(car.brand)
print(car.model)

car.start()
```

### Python Dunder Methods

### Definition

**Dunder methods** (short for "double underscore" methods) are special methods in Python whose names begin and end with two underscores, such as `__init__` and `__str__`.

They are also called **magic methods** or **special methods**. Dunder methods define how objects behave when they interact with Python's built-in operators, functions, and syntax.

---

### Why It Matters

Custom objects do not automatically behave like built-in Python types. Dunder methods allow us to define how objects should behave when used with Python's built-in syntax and functions.

They allow custom classes to:

- **Integrate with Python Syntax**: Define behavior for operators such as `+`, `-`, and `==`.
- **Work with Built-in Functions**: Define behavior for functions such as `len()`, `str()`, and `repr()`.
- **Provide Readable Representations**: Control how objects are displayed when printed or inspected.
- **Initialize Objects**: Automatically set an object's initial state when it is created.
- **Enable Pythonic Behavior**: Make custom objects behave similarly to built-in types such as lists and dictionaries.

---

### Key Dunder Methods

#### 1. Initialization: `__init__`

The `__init__` method is an **initializer method** that Python automatically calls after a new object is created.

- **Purpose**: Initializes the object's attributes and initial state.
- **Implicit Argument**: Receives `self` automatically as its first parameter.


> **Note:** `__init__` initializes an object; it does not actually create the object. Object creation is handled by `__new__`.

---

#### 2. String Representation: `__str__`

The `__str__` method defines the **human-readable string representation** of an object.

* **Trigger**: Automatically called by `str()` and `print()`.
* **Purpose**: Provides meaningful output instead of Python's default object representation.

Without `__str__`, `print(object)` would display Python's default object representation.

---

### Method Argument Types

When working with instance methods, arguments can be understood based on how they are passed to the method:

* **Implicit Arguments**: Arguments automatically passed by Python. For instance methods, `self` is automatically passed when the method is called through an object.
* **Explicit Arguments**: Arguments explicitly provided by the caller, such as `name` and `age`.

---

### Common Best Practices

* **Required Attributes**: Make essential attributes required during initialization when every object needs them.
* **Automated Identifiers**: Generate system-managed values automatically instead of requiring users to provide them.
* **Return Requirements**: `__str__` must return a string. Returning another data type causes a `TypeError`.

---

### Common Mistakes

* **Confusing `__init__` with Object Creation**: `__init__` initializes an object after it has been created. `__new__` is responsible for creating the instance.

* **Calling Dunder Methods Directly**: Prefer Python's built-in interfaces such as `str(obj)` instead of explicitly calling `obj.__str__()`.

* **Forgetting `self`**: Instance methods must define `self` as their first parameter so Python can pass the instance automatically.

---

### Quick Recap

* **Dunder methods** define how custom objects interact with Python's built-in syntax and functions.
* `__init__` **initializes** an object's attributes.
* `__str__` defines an object's **human-readable representation**.
* `self` is passed **implicitly** by Python when an instance method is called.
* Other arguments, such as `name` and `age`, are passed **explicitly** by the caller.
* Dunder methods make custom classes behave more like Python's built-in types.


```python
# Using __init__ method to initialize instance attributes
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p = Point(3, 4)
print(p.x)  
print(p.y)  

print(type(p)) # type() func returns the class an obj belongs to
```

```python
# Creating a class with a method to calculate the distance from the origin
class Point:
    def __init__(self, x, y): # Here, x and y are explicit arg
        self.x = x
        self.y = y
    def distance_from_origin(self): # Here, self is an implicit arg
        return ((self.x **2) +(self.y **2)) **0.5
    
point = Point(4,3)
print(point.distance_from_origin())
```

**NOTE** : When calling **attributes** of an instance, we use the dot notation `instance.attribute` and dont specify parentheses, whereas when calling **methods** of an instance, we use the dot notation `instance.method()` and specify parentheses to indicate that we are calling a function.

```python
# Converting an instance to a string using __str__ method
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __str__(self):
        return (f"Point({self.x}, {self.y})")
point_1 = Point(2, 5)
point_2 = Point(3, 6)
point_3 = Point(0,0)

del(point_2) # del() func deletes an object or attr of an obj
del(point_3.x)  

print(point_1)  # Now python will call the __str__ method to get the string representation of the object
print(point_3.y)

# print(point_2) will result in an error as point_2 doesnt exist
# print(point_3) will result in an error as point_3 obj has no attr 'x'
```

```python
# Instances as return values from methods
class Circle:
    def __init__(self, radius): # Here, radius is a positional arg
        self.radius = radius
    def area(self):
        return 3.14 * (self.radius ** 2)
    def circumference(self):
        return 2 * 3.14 * self.radius
print(Circle(5).area())
print(Circle(5).circumference())
```

```python
# Creating a Rectangle Class with area() and perimeter() instance methods
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


rectangle = Rectangle(10, 5)

print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
```

```python
# Creating a SMS_Store using OOP
class SMS_store:
    def __init__(self):
        self.inbox = []
    def __str__(self):
        return (f"{self.inbox}")
    def add_new_arrival(self, from_number, time_arrived, text_of_SMS):
        has_been_viewed = False
        message = (
            has_been_viewed,
            from_number,
            time_arrived,
            text_of_SMS
    )
        self.inbox.append(message)
    def message_count(self):
        total_messages = len(self.inbox)
        return total_messages
    def get_unread_indexes(self):
        unread_indexes = []
        for index, item in enumerate(self.inbox):
            if item[0] == False:
                unread_indexes.append(index)
        return unread_indexes
    def get_message(self, i):
        if i < len(self.inbox):
            message = self.inbox[i]
            self.inbox[i] = (True, message[1], message[2], message[3])
            return (message[1], message[2], message[3])
        else:
            return None
    def delete(self, i):
        if i < len(self.inbox):
            del self.inbox[i]
    def clear(self):    
        self.inbox.clear()

my_inbox = SMS_store()
my_inbox.add_new_arrival("0312", "10 PM", "Hello")
my_inbox.add_new_arrival("0322", "12 PM", "We will be coming late")
my_inbox.add_new_arrival("0314", "1 AM", "Open the door")
print(my_inbox)
print("Total Messages in inbox :",my_inbox.message_count())

print(my_inbox.get_message(0))
print(my_inbox)
print("Total Messages in inbox :",my_inbox.message_count())

print("Indexes of unread messages in inbox :",my_inbox.get_unread_indexes())

my_inbox.delete(0)
print(my_inbox)

my_inbox.add_new_arrival("0321", "11 PM", "Where are you?")
print(my_inbox)

my_inbox.clear()
print(my_inbox)
```
