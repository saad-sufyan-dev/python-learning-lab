# Introduction to strings and string methods

## What is a string?

A string is a sequence of characters used to represent text. In Python, strings are written inside single quotes (`' '`), double quotes (`" "`), or triple quotes (`''' '''` or `""" """`).

A string can contain:

- Letters: `"Python"`
- Numbers represented as text: `"123"`
- Special characters: `"@#$%"`
- Spaces: `"Hello World"`
- A combination of these: `"Saad123!"`

Important: A number inside quotation marks is a string, not a numeric value. For example, `"123"` is text, whereas `123` is an integer. Their types determine which operations can be performed on them.

## Creating strings

Python supports different quotation styles for defining strings.

| Quotation style | Common use                                                     |
| --------------- | -------------------------------------------------------------- |
| Single quotes   | Short strings                                                  |
| Double quotes   | Short strings, especially when the text contains an apostrophe |
| Triple quotes   | Multiline strings or text containing multiple lines            |

Single and double quotes generally serve the same purpose. Choosing between them often depends on which makes the text easier to write.

Triple-quoted strings can span multiple lines, making them useful for longer text. They can also be used for documentation strings, known as docstrings, when placed in appropriate locations.

## String indexing and slicing

Strings are ordered sequences, meaning each character has a position called an index.

Python uses zero-based indexing, so the first character has index `0`, the second has index `1`, and so on.

Consider the string `"PYTHON"`:

| Character      | `P` | `Y` | `T` | `H` | `O` | `N` |
| -------------- | --- | --- | --- | --- | --- | --- |
| Positive index | 0   | 1   | 2   | 3   | 4   | 5   |
| Negative index | -6  | -5  | -4  | -3  | -2  | -1  |

- Positive indexing counts from left to right, starting at `0`.
- Negative indexing counts from right to left, starting at `-1`.

### Slicing

Slicing extracts a portion of a string using a starting index and an ending boundary.

The general notation is `string[start:stop:step]`.

- `start`: Index where extraction begins; included in the result.
- `stop`: Index where extraction ends; excluded from the result.
- `step`: Number of positions to move between selected characters. Its default is `1`.

For example, slicing `"PYTHON"` from index `1` up to, but not including, index `4` selects `"YTH"`.

Slicing is useful for extracting parts of text, reversing strings, and selecting characters at regular intervals.

## What are string methods?

A string method is a built-in operation associated with the string type that performs a particular task, such as changing letter case, searching for text, or splitting a sentence.

Methods are accessed using dot notation. They operate on a string and return a result according to the method's behavior.

Python strings are immutable, meaning their contents cannot be changed after creation. Consequently, methods that appear to modify a string usually return a new string instead of changing the original.

For example, converting a string to uppercase produces a new uppercase string; the original string remains unchanged unless the returned value is assigned or used elsewhere.

## Common string methods

### 1. Changing letter case

These methods return strings with different capitalization.

| Method         | Purpose                                                                          |
| -------------- | -------------------------------------------------------------------------------- |
| `upper()`      | Converts letters to uppercase.                                                   |
| `lower()`      | Converts letters to lowercase.                                                   |
| `capitalize()` | Capitalizes the first character and lowercases the remaining characters.         |
| `title()`      | Capitalizes the beginning of each word according to Python's title-casing rules. |
| `swapcase()`   | Converts uppercase letters to lowercase and lowercase letters to uppercase.      |

Practical use: Case conversion is useful when comparing user input without treating uppercase and lowercase letters as different.

For instance, `"PYTHON"` and `"python"` are different strings when compared directly, but their lowercase forms are equal.

### 2. Removing whitespace and characters

Whitespace includes spaces, tabs, and newline characters.

| Method     | Purpose                                                                        |
| ---------- | ------------------------------------------------------------------------------ |
| `strip()`  | Removes specified characters from both ends; whitespace is removed by default. |
| `lstrip()` | Removes specified characters from the left end.                                |
| `rstrip()` | Removes specified characters from the right end.                               |

These methods do not remove matching characters from the middle of a string.

An important detail is that the optional argument to `strip()`, `lstrip()`, and `rstrip()` is treated as a set of characters, not as an exact substring.

For example, stripping the characters `a` and `b` from the ends removes any combination of those characters found at either end, stopping when another character is encountered.

### 3. Searching and counting

These methods help locate text within a string.

| Method         | Purpose                                                                                  |
| -------------- | ---------------------------------------------------------------------------------------- |
| `find()`       | Returns the index of the first occurrence of a substring, or `-1` if it is not found.    |
| `index()`      | Returns the index of the first occurrence, but raises a `ValueError` if it is not found. |
| `count()`      | Counts non-overlapping occurrences of a substring.                                       |
| `startswith()` | Checks whether a string begins with a specified prefix.                                  |
| `endswith()`   | Checks whether a string ends with a specified suffix.                                    |

Key distinction: `find()` and `index()` both search for text, but they handle unsuccessful searches differently. Use `find()` when a missing match is an expected possibility; use `index()` when the absence of a match should raise an error.

The methods `startswith()` and `endswith()` return Boolean values: `True` or `False`.

### 4. Replacing and splitting strings

| Method        | Purpose                                                                                            |
| ------------- | -------------------------------------------------------------------------------------------------- |
| `replace()`   | Returns a string with occurrences of one substring replaced by another.                            |
| `split()`     | Divides a string into a list of substrings.                                                        |
| `rsplit()`    | Splits from the right side of the string.                                                          |
| `partition()` | Divides a string into three parts: before the first separator, the separator itself, and after it. |
| `join()`      | Combines strings from an iterable using the calling string as a separator.                         |

For example, splitting a sentence at spaces separates its words into a list. Joining those words with a hyphen as the separator combines them into a single hyphen-separated string.

Important: `split()` returns a list, not another string. In contrast, `replace()` returns a string.

By default, `split()` treats consecutive whitespace as a single separator and ignores leading and trailing whitespace. Supplying an explicit separator changes this behavior.

### 5. Checking string contents

These methods return `True` or `False` depending on the characters in a string.

| Method        | Returns `True` when...                                                                             |
| ------------- | -------------------------------------------------------------------------------------------------- |
| `isalpha()`   | All characters are alphabetic and the string is non-empty.                                         |
| `isdigit()`   | All characters are digits recognized by Python's digit classification and the string is non-empty. |
| `isdecimal()` | All characters are decimal characters and the string is non-empty.                                 |
| `isalnum()`   | All characters are alphabetic or numeric and the string is non-empty.                              |
| `isspace()`   | All characters are whitespace and the string is non-empty.                                         |
| `islower()`   | All cased characters are lowercase, and at least one cased character exists.                       |
| `isupper()`   | All cased characters are uppercase, and at least one cased character exists.                       |

An empty string returns `False` for these checks.

The distinction between `isdigit()` and `isdecimal()` matters when working with Unicode characters. `isdigit()` accepts some digit-like characters that `isdecimal()` does not. Neither method is a complete replacement for validating a number according to every possible input format.

For example, these methods alone do not validate signed numbers, decimal numbers containing a decimal point, or formatted values containing commas.

### 6. Formatting strings

String formatting helps combine text with values, such as names, prices, or calculated results.

Python supports several approaches:

- F-strings: Embed expressions inside string literals prefixed with `f`.
- `format()`: Insert values into replacement fields within a string.
- Percent formatting: An older formatting approach that is still supported.

F-strings are commonly preferred in modern Python because they are readable and allow expressions to be included directly in the formatted text.

Formatting creates a string representation of values; it does not necessarily preserve their original data types.

## Common mistakes

- Forgetting that indexing starts at zero: The first character is at index `0`, not `1`.
- Expecting the stop index to be included: The stop boundary in slicing is excluded.
- Trying to change a string in place: Strings are immutable, so methods return new results.
- Confusing `find()` with `index()`: The former returns `-1` for a missing substring, while the latter raises an exception.
- Assuming `strip()` removes an exact word: Its character argument represents a set of characters, not a complete substring.
- Treating every numeric-looking string as a number: Text such as `"25"` must be converted to an appropriate numeric type before numerical calculations.
- Forgetting that methods return different types: For example, `split()` returns a list, while `upper()` returns a string.

## Quick recap

- A string represents text and is an immutable, ordered sequence of characters.
- Indexing accesses individual characters, while slicing extracts portions of a string.
- String methods perform operations such as case conversion, searching, replacing, splitting, and validation.
- Methods generally return new results rather than modifying the original string.
- Understanding return types and edge cases helps you choose the correct method for each task.

```python
# String Slicing

name = "Saad Sufyan"
print(name[0:4]) # Starts from index 0 and goes up to index 4 (not including 4)

print(name[-1:-7:-1]) # Starts from index -1 and goes up to index -7 (not including -7) with a step of -1
print(name[-6:len(name)]) # Starts from index -6 and goes up to the length of the string (which is last index+1 so it includes the last index)
```

```python
print(name[:4]) # is same as print(name[0:4])
print(name[0:4])

print(name[5:]) # is same as print(name[5:len(name)])
print(name[5:])
```

```python
a = "Harry is a  good  boy  "

b = a.find("  ") # Detects the first occurrence of double space
print(b)

c = a.replace("  ", " ") # Replaces all occurrences of double space with single space
c = c.strip() # Removes leading and trailing spaces
print(c)
```

```python
# isdecimal() vs isdigit()

# isdigit() is used for unicode characters, such as superscript numbers (², ³, ¹), subscripts (₀, ₁, ₂), decimal numbers (½, ¼),  and other numeric characters
# isdecimal() is used for decimal numbers (0-9) only

a = "10.1"
print(a.isdecimal()) # False - contains a decimal point
print(a.isdigit()) # False - contains a decimal point

b = "10"
print(b.isdecimal()) # True - contains only decimal numbers
print(b.isdigit()) # True - contains only decimal numbers

c = "²"
print(c.isdecimal()) # False - contains unicode character
print(c.isdigit()) # True - contains unicode character

d = "abc123"
print(d.isalpha()) # False - contains alphanumeric characters
print(d.isalnum()) # True - contains only alphanumeric characters

e = "abc 123"
print(e.isalpha()) # False - contains alphanumeric characters and a space
print(e.isalnum()) # False - contains alphanumeric characters and a space

f = "   "
print(f.isspace()) # True - contains only whitespace characters
```

```python
# Formatting strings

name = "John"
age = 30

# f-string formatting (modern and recommended)
print(f"Hello, {name}. You are {age} years old.")

# .format() method (more flexible than f-strings because it can be used with variables, expressions, etc.)
print("Hello, {}. You are {} years old.".format(name, age))

# % formatting (used in older Python versions, C-style, not recommended for new code)
print("Hello, %s. You are %d years old." % (name, age))
```

```python
# Escape Sequences
a = "Dear Harry,\n\tThis Python course is nice.\nThanks!" # \n is treated as a new line
print(a)

b = "My name is \"Saad Sufyan\"" # \" is treated as a double quote
print(b)

c = "This is a backslash: \\" # \\ is treated as a backslash
print(c)

d = "\'Hello\'" # \' is treated as a single character, not double
print(len(d))
```

```python
original_message = "  ALERT: Saad has won 50000 PKR!!! Claim NOW at www.example.com  "
cleaned_message = original_message.strip()
cleaned_message = cleaned_message[:7] + cleaned_message[7:-16].title() + cleaned_message[-16:]
cleaned_message = cleaned_message.replace("!", ".")
cleaned_message = cleaned_message.replace("50000", "100000")


upper_case = cleaned_message.upper()
lower_case = cleaned_message.lower()

print(f"'a' letter appears {lower_case.count("a")} times in lower_case")
print(f"'!' letter appears {cleaned_message.count("!")} times in cleaned_message")
print()

print(f"Message starts with 'ALERT': {cleaned_message.startswith("ALERT")}")
print(f"Message ends with 'com': {cleaned_message.endswith("com")}")
print()

modified_message = original_message.replace("50000", "100000")
print(f"Modified message: {modified_message}")
print()

print(f"Starting index of 'www.example.com': {cleaned_message.find("www.example.com")}")
print(f"Website URL: {cleaned_message[-16:]}")
print(f"'Claim NOW' in message: {'Claim NOW' in cleaned_message}")
print()

print(f"Length of cleaned_message: {len(cleaned_message)}")
print(f"Length of original_message: {len(original_message)}")

print(f"Cleaned message: {cleaned_message}")
print(f"Original message: {original_message}")
```
