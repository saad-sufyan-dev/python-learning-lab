# Introduction to Dictionaries and Sets in Python

Python provides several built-in data structures for organizing data. Two useful ones are dictionaries (`dict`) and sets (`set`).

## 1. Dictionaries

### What is a dictionary?

A dictionary is a mutable data structure that stores information as key-value pairs. Each key identifies its associated value.

For example, a student record might associate `name` with `"Saad"`, `age` with `15`, and `grade` with `10`.

### Key characteristics

- Unique keys: Duplicate keys are not stored separately; the last assigned value is retained.
- Any value type: Values can be strings, numbers, lists, or even other dictionaries.
- Hashable keys: Keys must be hashable, such as strings, integers, or suitable tuples.
- Mutable: Items can be added, modified, or removed.
- Insertion order: Dictionaries preserve the order in which keys were inserted.

### Accessing and modifying values

Dictionary values are accessed using their keys rather than numerical indexes.

- `dict[key]` retrieves a value but raises `KeyError` if the key is missing.
- `get()` retrieves a value safely, returning `None` or a specified default if the key is absent.
- Add or update: Associate a value with a new or existing key.
- `pop()` removes a specified key and returns its value.
- `clear()` removes all items.

### Important dictionary methods

| Method     | Purpose                                            |
| ---------- | -------------------------------------------------- |
| `keys()`   | Returns a view of all keys.                        |
| `values()` | Returns a view of all values.                      |
| `items()`  | Returns a view of key-value pairs.                 |
| `get()`    | Retrieves a value with a default for missing keys. |
| `update()` | Adds or updates key-value pairs.                   |
| `pop()`    | Removes a specified key and returns its value.     |
| `copy()`   | Creates a shallow copy.                            |

Use dictionaries when you need to associate information, such as product IDs with product names or student names with grades.

## 2. Sets

### What is a set?

A set is a mutable collection of unique, hashable elements. It is particularly useful for removing duplicates, checking membership, and comparing collections.

### Key characteristics

- Unique elements: Duplicate values are stored only once.
- Unordered: Sets do not guarantee iteration order and do not support positional indexing.
- Hashable elements: Elements must be hashable, such as strings, integers, or suitable tuples.
- Mutable: Elements can be added or removed.

### Creating sets

Sets use curly braces containing elements. However, empty curly braces (`{}`) create an empty dictionary, not a set. To create an empty set, use `set()`.

### Important set methods

| Method      | Purpose                                                           |
| ----------- | ----------------------------------------------------------------- |
| `add()`     | Adds one element.                                                 |
| `update()`  | Adds elements from another collection or iterable.                |
| `remove()`  | Removes an element; raises `KeyError` if absent.                  |
| `discard()` | Removes an element if present; does not raise an error if absent. |
| `pop()`     | Removes and returns an arbitrary element.                         |
| `clear()`   | Removes all elements.                                             |

### Set operations

Suppose set A contains students enrolled in Python, while set B contains students enrolled in SQL.

| Operation                      | Meaning                              | Example use                             |
| ------------------------------ | ------------------------------------ | --------------------------------------- |
| Union (`A \\| B`)              | All unique elements from both sets   | Students enrolled in either course      |
| Intersection (`A & B`)         | Elements common to both sets         | Students enrolled in both courses       |
| Difference (`A - B`)           | Elements in A but not in B           | Python students not enrolled in SQL     |
| Symmetric difference (`A ^ B`) | Elements in either set, but not both | Students enrolled in exactly one course |

Other useful methods include `issubset()`, `issuperset()`, and `isdisjoint()`, which check subset relationships, superset relationships, and whether two sets have no elements in common.

## 3. Dictionaries vs. Sets

| Feature    | Dictionary (`dict`)                | Set (`set`)              |
| ---------- | ---------------------------------- | ------------------------ |
| Purpose    | Store key-value pairs              | Store unique elements    |
| Duplicates | Keys are unique; values may repeat | Elements are unique      |
| Ordering   | Preserves insertion order          | No guaranteed order      |
| Access     | By key                             | By membership, not index |
| Empty form | `{}`                               | `set()`                  |

Both use hashing for efficient lookups. Dictionary keys and set elements must be hashable, but dictionary values can be of any type.

## 4. Common mistakes

- Confusing empty sets and dictionaries: `{}` creates a dictionary; `set()` creates an empty set.
- Assuming dictionaries use numerical indexes: Dictionary access uses keys, not positional indexes.
- Expecting duplicate set elements: Sets automatically retain only unique elements.
- Using unhashable elements: Lists and dictionaries cannot be dictionary keys or set elements.
- Confusing `remove()` and `discard()`: `remove()` raises `KeyError` when an element is missing; `discard()` does not.
- Assuming set order: Never rely on a set's iteration order or use positional indexing.
- Reversing set differences: `A - B` and `B - A` can produce different results.

## 5. Quick recap

- Dictionary: Stores key-value pairs; use it when information needs to be associated with a key.
- Set: Stores unique elements; use it for deduplication, membership checks, and comparisons.
- Dictionary keys and set elements must be hashable.
- Dictionaries preserve insertion order; sets do not guarantee an order.
- Union, intersection, difference, and symmetric difference help compare sets.
- Remember: `{}` is an empty dictionary, while `set()` is an empty set.

### Dictionaries

```python
# Dictionaries are mutable key-value pairs
# They are unordered (meaning they are not indexed and we access values by their keys)

marks = {
    "Maths": 90,
    "Science": 85,
    "English": 95
}

print(marks, type(marks))
print(marks["Maths"])

# Empty dictionary
empty_dict = {}
print(empty_dict, type(empty_dict))
```

```python
# The keys of a dictionary must be immutable objects, like strings, numbers, or tuples.
# whereas the values of them can be of any data type, even other dictionaries

dict = {
    "name": "Fahad",
    10 : True,
    (1, 2, 3) : "tuple_key",
    "marks" : [85, 90, 78],
    "record" : {
        "class": "10th",
        "section": "A"
    }
}

print(dict[(1, 2, 3)], type((1, 2, 3)))
print(dict["record"]["class"], type(dict["record"]))
```

```python
# Dictionary Methods

print(dict.keys()) # Returns a view object of all keys

print(dict.values()) # Returns a view object of all values

print(dict.items()) # Returns a view object of all key-value pairs (in list format)

print(dict.get("record", "Not found")) # Returns the value for the specified key, or the default value if the key is not found

# Updates the dictionary with the specified key-value pair, if already exists, 
# it will update the value, otherwise it will create a new key-value pair
dict["record"].update({"section": "D"}) 

print(dict.get("record", {}).get("section")) # getting the value of the key "section" from the dictionary "record" (key of dict)

# Returns the value for the specified key, but if value doesn't exist, 
# it will create a new key-value pair and insert the default value
print(dict.setdefault("Group", "CS"), dict) 

dict.pop((1,2,3)) # Removes the key-value pair with the specified key
print(dict)
```

**view object** is a **dynamic reference** to the elements of the dictionary, meaning it updates automatically when the dictionary changes.

```python
student = {
    "name": "Saad",
    "age": 15,
    "subjects": ("Math", "Physics", "Chemistry"),
    "marks": {
        "Math": 87,
        "Physics": 92,
        "Chemistry": 81
    },
    "city": "Karachi"
}

print(f"Name of student is : {student.get('name', 'No Value')}")
print(f"Age of student is : {student.get('age', 'No Value')}")
print(f"City of student is : {student.get('city', 'No Value')}")
print()

total_marks = sum(student["marks"].values())
avg_marks = total_marks / len(student["marks"])

student["record"] = {}
student["record"]["total_marks"] = total_marks
student["record"]["avg_marks"] = avg_marks

print(f"Total Marks before adding Computer Science: {student['record']['total_marks']}")
print(f"Avg Marks before adding Computer Science: {student['record']['avg_marks']}")

student["marks"].update({"Computer Science": 85})

total_marks = sum(student["marks"].values())
avg_marks = total_marks / len(student["marks"])
student["record"].update({"total_marks": total_marks, "avg_marks": avg_marks})

print(f"Total marks after adding Computer Science: {student['record']['total_marks']}")
print(f"Avg marks after adding Computer Science: {student['record']['avg_marks']}")
print()

student.setdefault("grade", "Not Assigned")

student.update({"city": "Lahore"})

backup = student.copy()
print(f"Shallow copy of student: {backup}")

backup.pop("age")
print(f"Backup after removing age: {backup}")
print(f"Student after removing age key from backup: {student}")
print()

print(f"Parantal Level of education: {student.get('parantal_education', 'Not Specified')}")
print()

print(f"Keys of student: {student.keys()}")
print(f"Values of student: {student.values()}")
print(f"Items of student: {student.items()}")
```

```python
inventory = {
    "Laptop": {
        "price": 150000,
        "stock": 5,
        "category": "Electronics"
    },
    "Phone": {
        "price": 80000,
        "stock": 10,
        "category": "Electronics"
    },
    "Desk": {
        "price": 25000,
        "stock": 4,
        "category": "Furniture"
    }
}

laptop_revenue = inventory["Laptop"]["price"] * inventory["Laptop"]["stock"]
phone_revenue = inventory["Phone"]["price"] * inventory["Phone"]["stock"]
desk_revenue = inventory["Desk"]["price"] * inventory["Desk"]["stock"]

total_potential_revenue = laptop_revenue + phone_revenue + desk_revenue
print(f"Total potential revenue: {total_potential_revenue}")

inventory.update({"Headphones": {"price": 5000, "stock": 20, "category": "Electronics"}})

inventory["Laptop"].setdefault("discount", 0)

inventory["Phone"].update({"stock": 15, "price": 75000})

backup = inventory.copy()
backup.pop("Desk")

laptop_revenue = inventory["Laptop"]["price"] * inventory["Laptop"]["stock"]
phone_revenue = inventory["Phone"]["price"] * inventory["Phone"]["stock"]
desk_revenue = inventory["Desk"]["price"] * inventory["Desk"]["stock"]
headphones_revenue = inventory["Headphones"]["price"] * inventory["Headphones"]["stock"]

total_potential_revenue = laptop_revenue + phone_revenue + desk_revenue + headphones_revenue
print(f"Updated Total potential revenue: {total_potential_revenue}")
print()

print(f"Price of Playstation: {inventory.get('Playstation', {}).get('price', 'Not found')}")
print()

fav = {
    "Laptop": inventory["Laptop"],
    "Desk": inventory["Desk"]
}
print(fav)
print()

print(f"Keys of inventory: {list(inventory.keys())}")
print(f"Values of inventory: {list(inventory.values())}")
print(f"Items (key-value pairs) of inventory: {list(inventory.items())}")
```

```python
# Deep Copy v/s Shallow Copy

import copy

# Shallow Copy
original_dict = {'a': 1, 'b': 2, 'c': [3, 4]}
shallow_copy = original_dict.copy() # copies the outer dictionary as a seperate object, but the inner list is still referenced

# Deep Copy
deep_copy = copy.deepcopy(original_dict) # copies the outer dictionary as a seperate object, and the inner list as well

print("Original Dict:", original_dict)
print("Shallow Copy:", shallow_copy)
print("Deep Copy:", deep_copy)

# Modify the inner list in the original dictionary
original_dict['c'].append(5)

print("\nAfter modifying the inner list in the original dictionary:")
print("Original Dict:", original_dict)
print("Shallow Copy:", shallow_copy) # will be affected, as the inner list is still referenced
print("Deep Copy:", deep_copy) # will not be affected, as the inner list is a separate object

# Modify the outer dictionary
original_dict['d'] = 6

print("\nAfter modifying the outer dictionary:")
print("Original Dict:", original_dict)
print("Shallow Copy:", shallow_copy) # will not be affected, as the outer dictionary is a separate object
print("Deep Copy:", deep_copy) # will not be affected, as the outer dictionary is a separate object
```
