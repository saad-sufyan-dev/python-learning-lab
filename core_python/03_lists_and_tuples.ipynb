# Introduction to Lists and Tuples

## 1. What are lists and tuples?

Lists and tuples are built-in Python data types used to store multiple items in a single variable. They are both sequence types, meaning their items are stored in a specific order and can be accessed using their positions, called indexes.

For example, a student's marks in five subjects can be stored together instead of creating five separate variables.

Lists and tuples can store different data types, including integers, strings, floats, and even other lists or tuples.

## 2. Lists

A list is an ordered collection of items that can be modified after creation. This property is called mutability.

Lists are useful when data may need to be added, removed, or updated, such as a shopping cart, a student's changing marks, or a collection of tasks.

### Key characteristics

- Ordered: Items maintain their sequence.
- Mutable: Items can be added, removed, or modified.
- Allows duplicates: The same value can appear multiple times.
- Allows mixed data types: A single list can contain integers, strings, and other objects.
- Indexed: Items can be accessed by their positions, starting from index `0`.

### Creating a list

Lists are created using square brackets, `[]`, with items separated by commas.

A list containing three subjects might hold `"Physics"`, `"Chemistry"`, and `"Mathematics"`. Each subject occupies a different index:

| Index | Item        |
| ----- | ----------- |
| `0`   | Physics     |
| `1`   | Chemistry   |
| `2`   | Mathematics |

The first item is at index `0`, not index `1`. This is called zero-based indexing.

### Modifying a list

Because lists are mutable, you can:

- Access an item using its index.
- Update an existing item.
- Append a new item to the end.
- Insert an item at a specified position.
- Remove an item by value or position.
- Sort items when their values support the required ordering.

Python provides methods such as `append()`, `insert()`, `remove()`, `pop()`, and `sort()` for these operations.

> Important: Changing an item in a list modifies the existing list rather than requiring you to create a new list.

## 3. Tuples

A tuple is an ordered collection of items that cannot be modified after creation. This property is called immutability.

Tuples are useful for representing collections of values that should remain fixed, such as a coordinate, an RGB color, or a date represented by its year, month, and day.

### Key characteristics

- Ordered: Items maintain their sequence.
- Immutable: Items cannot be added, removed, or replaced through ordinary item assignment after the tuple is created.
- Allows duplicates: Repeated values are permitted.
- Allows mixed data types: Items can have different types.
- Indexed: Items can be accessed using zero-based indexing.

### Creating a tuple

Tuples are commonly created using parentheses, `()`, with items separated by commas.

For example, a coordinate could contain the two values `10` and `20`. The first value represents the horizontal position, and the second represents the vertical position.

### The special case of a single-item tuple

A tuple containing only one item requires a trailing comma. Parentheses alone do not create a tuple; they can simply group an expression.

For example, `(5)` is an integer, whereas `(5,)` is a tuple.

The comma is what makes the difference.

### Why use tuples?

Tuples are useful when the collection itself should remain fixed. Their immutability can help prevent accidental changes and communicate that a group of values represents one fixed record.

However, immutability has an important detail: if a tuple contains a mutable object, such as a list, that inner object may still be modified. The tuple prevents replacing its elements, not necessarily changing the contents of every object it references.

## 4. Lists vs. tuples

| Feature              | List                    | Tuple                         |
| -------------------- | ----------------------- | ----------------------------- |
| Syntax               | Square brackets `[]`    | Parentheses `()` commonly     |
| Mutability           | Mutable                 | Immutable                     |
| Item assignment      | Allowed                 | Not allowed                   |
| Add or remove items  | Allowed                 | Not directly                  |
| Duplicates           | Allowed                 | Allowed                       |
| Mixed data types     | Allowed                 | Allowed                       |
| Indexing and slicing | Supported               | Supported                     |
| Common use           | Collections that change | Collections that remain fixed |

Both types support common sequence operations, including indexing, slicing, iteration, and checking whether an item exists.

Tuples are not automatically better than lists for performance. Choose based primarily on whether the collection should be mutable and what the data represents.

## 5. When should you use each?

Use a list when the collection needs to change.

Examples include:

- A shopping cart whose items can be added or removed.
- A to-do list whose tasks change.
- Student marks that may be updated.
- A collection of records being processed during data analysis.

Use a tuple when the collection should remain fixed.

Examples include:

- A two-dimensional coordinate.
- An RGB color represented by three values.
- A fixed set of configuration values.
- A record containing a person's year, month, and day of birth.

## 6. Important concepts to remember

- Indexing accesses an item by its position. Python sequences use zero-based indexing.
- Slicing retrieves a portion of a sequence without requiring the original sequence to be modified.
- Mutability determines whether an object can be changed after creation.
- Nested collections occur when a list or tuple contains other lists or tuples.
- Empty collections are possible: an empty list is `[]`, and an empty tuple is `()`.

One final distinction: assigning a list to another variable does not automatically create an independent copy. Both variables may refer to the same list object, so a modification through one variable can be visible through the other. Understanding this becomes important when working with mutable objects.

## Quick recap

- List: An ordered, mutable collection of items.
- Tuple: An ordered, immutable collection of items.
- Both support indexing, slicing, duplicates, and mixed data types.
- Lists are appropriate for changing collections; tuples are appropriate for fixed collections.
- A single-item tuple requires a trailing comma.
- Choosing between them depends on the intended behavior of the data, not merely on their syntax.

```python
# Lists can contain any data type, including other lists
# Lists are mutable, meaning they can be changed after creation
list = [1, "apple", 3.14, True, [1, 2, 3]]

print("Original list:", list)
print("Length of list:", len(list))

list.append("banana") # Modifies the original list, and no new object is created (like strings)

print("After appending 'banana':", list)
```

```python
# Lists methods

a = [1, 2, 3, 4, 5]
a.sort() # Sorts the list in ascending order
print(a)

b = [5, 4, 3, 2, 1]
b.reverse() # Reverses the list
print(b)
b.sort(reverse=True) # Sorts the list in descending order
print(b)
```
