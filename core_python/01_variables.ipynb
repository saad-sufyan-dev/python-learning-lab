# Variables

## What is a Variable?

A **variable** is a name that refers to a value stored or managed by a program.

Think of a variable as a **label attached to a piece of information**.

For example, a program might need to remember:

* A student's age
* A person's name
* A product's price
* The number of items in a cart

Instead of repeatedly writing the actual value, the program can give it a meaningful name.

---

## Creating a Variable

Python uses the **assignment operator** `=` to associate a variable name with a value.

Conceptually:

> **variable name → value**

For example:

`age` → `15`

Here:

* `age` is the **variable name**
* `15` is the **value**
* `=` performs the **assignment**

Assignment does **not** mean mathematical equality.

It means:

> "Associate this name with this value."

---

## Variables Can Refer to Different Types of Data

A variable does not have to contain only numbers.

| Data type | Example value | Meaning        |
| --------- | ------------- | -------------- |
| `int`     | `15`          | Whole number   |
| `float`   | `72.5`        | Decimal number |
| `str`     | `"Python"`    | Text           |
| `bool`    | `True`        | Boolean value  |

The type describes **what kind of value** the variable refers to.

Python is **dynamically typed**, meaning you do not normally have to explicitly declare a variable's type before assigning a value.

---

## Changing a Variable

A variable can be reassigned to a different value.

For example, imagine a variable representing a student's score:

`score` → `80`

After reassignment:

`score` → `95`

The variable name remains `score`, but it now refers to a different value.

This is why variables are useful: they allow programs to work with information that can change while the program runs.

---

## Variable Names

A good variable name should describe the information it represents.

Good examples:

* `age`
* `student_name`
* `total_price`
* `average_score`

Poor examples:

* `x`
* `a1`
* `thing`

Short names such as `x` can be appropriate in mathematical formulas or very small contexts, but descriptive names generally make programs easier to understand.

### Basic naming rules

Python variable names:

* Can contain letters, numbers, and underscores (`_`)
* Cannot start with a number
* Are **case-sensitive**
* Cannot use Python keywords such as `if`, `for`, or `class`

For example, `age`, `Age`, and `AGE` are three different names.

---

## Variable vs. Value

This distinction is important:

> A **variable is a name**; a **value is the data** associated with that name.

Suppose:

`price` → `250`

Then:

* `price` is the variable
* `250` is the value
* `int` is the value's data type

This distinction becomes increasingly important when learning data structures, functions, and object-oriented programming.

---

## Variables in a Program

Variables allow a program to **store information, use that information, and update it**.

A simple conceptual flow is:

**Input → Store in variables → Process → Produce output**

For example, a shopping program might store:

* `product_price`
* `quantity`
* `discount`

It can then use those values to calculate the final price.

The variables act as the program's **named pieces of information**.

---

## Common Mistakes

### 1. Confusing `=` with equality

In Python, `=` is primarily used for **assignment**.

Mathematical equality and comparison use a different operator: `==`.

### 2. Using invalid names

A variable name cannot begin with a number.

For example, a name beginning with `2` is invalid.

### 3. Ignoring case sensitivity

`name` and `Name` are different variable names.

### 4. Using unclear names

A name such as `total_sales` communicates much more information than `x`.

---

## Quick Recap

* **Python** is a high-level, general-purpose programming language.
* A **variable** is a name that refers to a value.
* The `=` operator performs **assignment**.
* Variables can refer to different data types such as `int`, `float`, `str`, and `bool`.
* Variables can be **reassigned** to different values.
* Python is **dynamically typed**.
* Good variable names are descriptive and follow Python's naming rules.
* Remember the fundamental distinction: **variable = name, value = data**.

# Python Operators

**Operators** are symbols or keywords used to perform operations on values and variables.

For example, addition uses the `+` operator, comparison uses `>`, and logical operations use `and`.

Python operators can be grouped into several categories.

## 1. Arithmetic operators

Arithmetic operators perform mathematical calculations.

| Operator | Name           | Meaning                                        |
| -------- | -------------- | ---------------------------------------------- |
| `+`      | Addition       | Adds two values                                |
| `-`      | Subtraction    | Subtracts the right value from the left        |
| `*`      | Multiplication | Multiplies two values                          |
| `/`      | Division       | Divides and returns a `float`                  |
| `//`     | Floor division | Divides and rounds down to the nearest integer |
| `%`      | Modulus        | Returns the remainder                          |
| `**`     | Exponentiation | Raises a value to a power                      |

For example, if `a = 10` and `b = 3`:

* `a + b` → `13`
* `a - b` → `7`
* `a * b` → `30`
* `a / b` → `3.333...`
* `a // b` → `3`
* `a % b` → `1`
* `a ** b` → `1000`

### `/` vs `//`

This is an important distinction:

* `/` performs **true division** and normally produces a floating-point result.
* `//` performs **floor division**, meaning the result is rounded **down**.

For example, `7 // 2` gives `3`.

> **Important:** Floor division means rounding toward negative infinity, not simply removing the decimal part. Therefore, `-7 // 2` gives `-4`.

---

## 2. Comparison operators

Comparison operators compare two values and produce a **Boolean** result: `True` or `False`.

| Operator | Meaning                  |
| -------- | ------------------------ |
| `==`     | Equal to                 |
| `!=`     | Not equal to             |
| `>`      | Greater than             |
| `<`      | Less than                |
| `>=`     | Greater than or equal to |
| `<=`     | Less than or equal to    |

For example, if `age = 18`:

* `age == 18` → `True`
* `age > 15` → `True`
* `age < 15` → `False`
* `age != 20` → `True`

### `=` vs `==`

This is one of the most common beginner mistakes:

* `=` is the **assignment operator**. It assigns a value to a variable.
* `==` is the **comparison operator**. It checks whether two values are equal.

---

## 3. Logical operators

Logical operators combine or modify Boolean conditions.

| Operator | Meaning                                  |
| -------- | ---------------------------------------- |
| `and`    | `True` if both conditions are true       |
| `or`     | `True` if at least one condition is true |
| `not`    | Reverses the Boolean result              |

Suppose:

* condition A = `True`
* condition B = `False`

Then:

* `A and B` → `False`
* `A or B` → `True`
* `not A` → `False`

Logical operators are especially useful when creating conditions.

For example, an eligibility rule might require someone to be at least 18 **and** have a valid ID.

---

## 4. Assignment operators

Assignment operators assign values to variables.

The basic assignment operator is:

* `=` → assigns a value

Python also provides **augmented assignment operators**, which combine an arithmetic operation with assignment.

| Operator | Equivalent idea         |
| -------- | ----------------------- |
| `=`      | Assign                  |
| `+=`     | Add and assign          |
| `-=`     | Subtract and assign     |
| `*=`     | Multiply and assign     |
| `/=`     | Divide and assign       |
| `//=`    | Floor-divide and assign |
| `%=`     | Modulus and assign      |
| `**=`    | Exponentiate and assign |

For example, if `x` initially contains `10`, then `x += 5` means: take the current value of `x`, add `5`, and store the result back in `x`.

So `x` becomes `15`.

---

## 5. Identity operators

Identity operators check whether two references point to the **same object**.

| Operator | Meaning             |
| -------- | ------------------- |
| `is`     | Same object         |
| `is not` | Not the same object |

This is different from `==`.

### `==` vs `is`

* `==` asks: **"Do these objects have equal values?"**
* `is` asks: **"Are these references pointing to the exact same object?"**

For normal value comparison, use `==`.

`is` is commonly used when checking for special singleton objects such as `None`.

---

## 6. Membership operators

Membership operators check whether a value exists inside a collection such as a string, list, tuple, set, or dictionary.

| Operator | Meaning               |
| -------- | --------------------- |
| `in`     | Exists inside         |
| `not in` | Does not exist inside |

For example, conceptually:

* `"a" in "apple"` → `True`
* `5 in [1, 2, 3]` → `False`

They are useful when you need to ask whether a particular value is present in a collection.

---

## 7. Bitwise operators

Bitwise operators work directly with the **individual binary bits** of integer values.

| Operator | Name        |
| -------- | ----------- |
| `&`      | Bitwise AND |
| `\|`     | Bitwise OR  |
| `^`      | Bitwise XOR |
| `~`      | Bitwise NOT |
| `<<`     | Left shift  |
| `>>`     | Right shift |

These are more advanced and are commonly encountered in areas such as:

* Low-level programming
* Networking
* Permissions and bit flags
* Performance-sensitive operations
* Cryptography-related programming

For normal beginner-level Python programming, arithmetic, comparison, logical, assignment, identity, and membership operators are more important to master first.

---

## Operator precedence

When an expression contains multiple operators, Python follows **operator precedence** to determine which operation happens first.

A simplified order is:

1. `()` — parentheses
2. `**` — exponentiation
3. `*`, `/`, `//`, `%`
4. `+`, `-`
5. Comparisons: `==`, `!=`, `<`, `>`, `<=`, `>=`
6. `not`
7. `and`
8. `or`

For example, multiplication is performed before addition.

Parentheses can be used when you want to explicitly control the order of evaluation.

> **Best practice:** When an expression could be confusing, use parentheses to make your intended order clear.

## Quick recap

Python operators can be grouped as:

* **Arithmetic** → mathematical calculations
* **Comparison** → compare values
* **Logical** → combine conditions
* **Assignment** → assign/update variables
* **Identity** → check whether objects are the same object
* **Membership** → check whether a value exists in a collection
* **Bitwise** → operate on binary bits

The most important distinction to remember early on is:

> `=` **assigns**, `==` **compares**, and `is` **checks object identity**.
