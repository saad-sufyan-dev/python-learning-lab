# Supervised Regression Learning Algorithms

## Definition

**Supervised regression** is a subfield of machine learning where an algorithm learns to map input features ($\mathbf{x}$) to a **continuous numerical output** ($y$) based on labeled historical data.

Unlike classification, which predicts discrete categorical labels, regression models estimate quantities on a continuous spectrum, such as prices, temperatures, stock values, or physical measurements.

---

## Why It Matters

Organizations rely on quantitative forecasting to make operational, strategic, and financial decisions. Rather than relying on guesswork, regression models allow data scientists to model relationships between independent variables (predictors) and dependent target metrics.

Supervised regression models enable applications such as:

* **Financial Forecasting**: Predicting future real estate values, stock prices, or sales revenue.
* **Risk & Underwriting**: Estimating insurance claim amounts or credit risk limits.
* **Resource Management**: Forecasting power consumption, cloud computing demand, or inventory needs.
* **Scientific Modeling**: Estimating biological growth rates, drug dosage responses, or weather conditions.

---

## Common Regression Algorithms

Different algorithms handle relationships between variables, dataset sizes, and noise in distinct ways. Selecting the appropriate algorithm depends on data structure, linearity, interpretability requirements, and computational resources.

### 1. Linear Regression

The fundamental baseline algorithm that models the target as a linear combination of input features.

* **Key Concept**: Fits a line (or hyperplane in higher dimensions) that minimizes the sum of squared residuals between predicted and actual values (Ordinary Least Squares).
* **When to Use**: When the relationship between features and target is predominantly linear, or when maximum model interpretability is required.
* **Limitations**: Highly sensitive to outliers, assumes linear feature independence, and fails on complex non-linear patterns.

### 2. Regularized Linear Models (Ridge, Lasso, ElasticNet)

Extensions of linear regression that add a **penalty term** to the loss function to prevent overfitting and manage multicollinearity.

* **Ridge Regression ($L_2$ Regularization)**: Adds a penalty proportional to the square of feature coefficients. Shrinks coefficients toward zero, reducing variance without eliminating features.
* **Lasso Regression ($L_1$ Regularization)**: Adds a penalty proportional to the absolute value of coefficients. Forces coefficients of irrelevant features to exactly zero, performing implicit feature selection.
* **ElasticNet**: Combines both $L_1$ and $L_2$ penalties, balancing feature selection with coefficient stabilization when features are correlated.
* **When to Use**: High-dimensional datasets (many features relative to samples) or when features exhibit high multicollinearity.

### 3. Decision Tree Regressor

A non-parametric model that partitions data into smaller regions based on feature split thresholds, predicting the average value of training samples in each leaf node.

* **Key Concept**: Uses metrics like **Variance Reduction** or **Mean Squared Error (MSE)** to find optimal split points in a tree hierarchy.
* **When to Use**: Capturing complex non-linear relationships and feature interactions without requiring feature scaling.
* **Limitations**: High risk of overfitting on training data; sensitive to small variations in data.

### 4. Ensemble Methods (Random Forest, Gradient Boosting)

Advanced algorithms that combine multiple decision trees to produce more robust predictions than a single model.

* **Random Forest (Bagging)**: Trains numerous decision trees in parallel on bootstrap subsets of the data and averages their individual predictions. Reduces variance and prevents overfitting.
* **Gradient Boosted Trees (Boosting - XGBoost, LightGBM, CatBoost)**: Builds trees sequentially, where each new tree is trained to correct the residual errors made by preceding trees.
* **When to Use**: Medium to large tabular datasets where high predictive performance is the primary objective.
* **Limitations**: Requires more computational resources and hyperparameter tuning; acts as a "black box" compared to simpler linear models.

### 5. Support Vector Regressor (SVR)

An adaptation of Support Vector Machines designed for continuous target prediction.

* **Key Concept**: Fits predictions within an $\epsilon$-insensitive margin (tube) around the regression function, ignoring errors that fall within this boundary. Uses kernel functions (e.g., RBF) to map non-linear relationships.
* **When to Use**: Small to medium-sized datasets with non-linear patterns or high-dimensional feature spaces.
* **Limitations**: Computationally expensive and slow to train on large datasets.

---

## Model Evaluation Metrics

Regression performance is evaluated by measuring the magnitude of prediction errors ($e_i = y_i - \hat{y}_i$).

| Metric | Full Name | Description & Use Case |
| --- | --- | --- |
| **MSE** | Mean Squared Error | Calculates the average of squared errors. Penalizes large errors heavily, making it useful when outliers are costly. |
| **RMSE** | Root Mean Squared Error | The square root of MSE. Expresses error in the **same unit as the target variable**, making interpretation straightforward. |
| **MAE** | Mean Absolute Error | Calculates the average of absolute errors. Treats all error sizes linearly and is robust against extreme outliers. |
| **$R^2$ Score** | Coefficient of Determination | Measures the proportion of variance in the target variable explained by the model (ranges from negative infinity to $1.0$). |

## Workflow for Regression Modeling

1. **Problem Formulation**: Confirm that the target variable is continuous and therefore suitable for a regression problem.

2. **Data Cleaning**: Handle duplicates, incorrect data types, irrelevant columns, missing values, and other data-quality issues.

3. **Exploratory Data Analysis (EDA)**: Inspect the target distribution, examine feature relationships and correlations, and identify potential outliers and patterns.

4. **Separate Features and Target**: Separate the feature variables (`X`) from the target variable (`y`).

5. **Data Splitting**: Use `train_test_split()` to partition the data into training and testing sets before fitting preprocessing transformations.

6. **Data Preprocessing**:

   * Handle missing values.
   * Scale numerical features when required.
   * Encode categorical features using One-Hot Encoding.
   * Fit preprocessing steps on the training data and apply the same transformations to the test data to prevent data leakage.

7. **Model Baseline & Selection**: Train **Linear Regression** as a baseline, then evaluate more complex models such as **Decision Tree** and **Random Forest**.

8. **Model Evaluation**: Evaluate each model using **MAE, RMSE, and R²** and compare their performance.

9. **Hyperparameter Tuning**: Optimize the best-performing model by tuning relevant hyperparameters using cross-validation.

10. **Final Model**: Select the best-performing model and retrain it using the appropriate training data and optimized hyperparameters.

11. **Final Evaluation**: Evaluate the final model once on the unseen test set using **MAE, RMSE, and R²**.

---

## Common Mistakes

* **Treating Regression as Classification**: Attempting to predict continuous quantities by grouping targets into arbitrary discrete buckets without an analytical reason.
* **Data Leakage during Preprocessing**: Scaling features, imputing missing values, or applying target-based transformations on the full dataset before performing train-test splits.
* **Ignoring Outliers in Linear Models**: Outliers disproportionately skew Ordinary Least Squares line fits due to squared error penalties.
* **Evaluating Trees with Out-of-Bound Extrapolation**: Tree-based models (Decision Trees, Random Forests) cannot extrapolate continuous values outside the range of training targets.
* **Relying Solely on $R^2$**: High $R^2$ values do not guarantee that a model is unbiased or predicting well on unseen scenarios; always inspect residual plots.

---

## Quick Recap

* **Supervised Regression** predicts continuous numerical values rather than discrete categorical labels.
* **Linear models** provide simple, interpretable baselines but require strict data assumptions.
* **Regularization (L1/L2)** prevents linear models from overfitting when handling complex or high-dimensional data.
* **Tree Ensembles (Random Forest, Gradient Boosting)** offer top performance on complex non-linear tabular data.
* **RMSE and MAE** quantify prediction error in target units, while **$R^2$** measures explained target variance.

### Correlation in Machine Learning



#### Definition

**Correlation** measures the strength and direction of a linear relationship between two numerical variables. It produces a score between **-1** and **+1**.

---

#### Why It Matters

* **Feature Selection**: Identifies redundant features that provide duplicate information.


* **Data Understanding**: Uncovers hidden relationships between variables.


* **Model Stability**: Helps prevent **multicollinearity**, which degrades linear model performance.



---

#### Correlation Values

* **+1**: Perfect positive relationship.


* **0**: No linear relationship.


* **-1**: Perfect negative relationship.



---

#### Important Consideration

> **Correlation does not imply causation**. Two correlated variables do not automatically cause each other to change.
> 
>

```python
# Inspecting correlations of features with the price column
corr = data.corr(numeric_only=True)["price"]
sns.heatmap(corr.to_frame(), annot=True, cmap="coolwarm")
```

```python
# Selecting features which have high correlation with price column
filtered_corr = corr[(corr>0.2) | (corr<-0.2)].drop("price")
sns.heatmap(filtered_corr.to_frame(), annot=True, cmap="coolwarm")
```
