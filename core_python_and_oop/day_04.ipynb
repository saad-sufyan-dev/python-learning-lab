# Exception Handling in python

Generally, when we execute a python program, we use a python interpreter (it translates code to binary line by line) as python is an interpreted  
language, unlike C and C++.  

When a python interpreter encounters an error while interpreting (such as converting a string into an integer), it returns to the program and  
checks if user has provided what to do if such error occured (how to encounter ValueError).  
If yes, it handles error accordingly, otherwise, it crashes and terminates flow of program.  

For Example, if we have 10 lines of code and error happens at line 5, then flow of program will terminate at line 5.  
Thats why its imp to learn how to handle exceptions in python.

````python
# Using try and except 
try:
    a = int(input("Enter an integer: "))
    print(f"Multiplication Table of {a} is:")
    for i in range(1, 11):
        print(f"{a} X {i} = {a * i}")
except ValueError:
    print("Input not an integer!")
except:
    print("Some error occured.")
    
print("\nSome imp lines of code.")
# If we enter a number, code will execute normally. But if someone enters a string, instead of crashing the program the interpreter would 
# handle that exception (printing that input is invalid) and successfully execute whole program without halting in the middle.
````

````python
# In python we can also raise errors and halt the program execution, like when a situation breaks real-life logic,
# even if it doesnt trigger a standard system crash. 

# Using raise keyword

def withdraw_money(balance, amount):
    if amount <= 0:
        raise ValueError("Withdrawal amount must be greater than zero.")
    
    if amount > balance:
        # Raising an error because you cannot withdraw more than you own
        raise PermissionError("Insufficient funds for this transaction.")
    
    return balance - amount

try:
    print(withdraw_money(5000, 6000))

except PermissionError as error:
    print(f"Transaction Denied: {error}")

except:
    print("Some error occured.")
````

```text
[ Code Execution ]
 └── Call: withdraw_money(5000, 6000)
      ├── Condition Check: 6000 > 5000 (True)
      └── raise PermissionError ──> (Execution Halts)
                                           │
[ Exception Handling ]                     ▼
 └── caught by 'except PermissionError as error'
      └── Prints: "Transaction Denied: Insufficient funds for this transaction."

````python
a = input("Enter a number: ")
try: 
    a = int(a)
    b = 1/a
    print(f"1 divided by {a} is {b}")

except ZeroDivisionError: 
    print("Cant divide by 0!")

except ValueError:
    print("Invalid input, not an integer")

except Exception as e: # here storing any other error in an alias 'e' and handling it using except 
    print(f"Some error occured:\n{e}")
````

```text
Syntax for try-except block:

try:  
    # risky code  
except:  
    # handle error  
else:
    # code when 'except' is skipped
finally:  
    # always runs

````python
# The finally clause in try-except block: it always runs whether an error occurs or not
# Common use-case: Closing a file

file = None
try:
    file = open("data.txt")
    data = file.read()
except FileNotFoundError:
    print("File not found")
finally:
    if file != None:
        file.close()

# Usually we use 'with' syntax, which automatically closes file
try:
    with open("data.txt") as f:
        data = f.read()

except FileNotFoundError:
    print("File not Found")
````

````python
# Similarly, you can have code that executes only if the try block succeeds using else clause
def attempt_int(n):
    try:
        int(n)
    except Exception as e:
        print(f"Failed: {e}")
    else:
        print("Succeeded")

attempt_int(2.3)
````

# Statements (break, continue and pass) in python

````python
# break statement terminates the loop completely and transfers flow of execution to the next statement outside the loop
# Usually used with for-else statement, where the else block belongs to for loop instead of if statement/
numbers = [1,2,3,4]

for num in numbers:
    if num == 4:
        print("Found")
        break # --> If val is found, 'break' will terminate the loop, and the else block will be skipped

else: # --> If no break statement executes during looping, then else block will execute 
    print("Not Found")
````

````python
# continue statement skips remaining code inside current iteration inside the loop and transfers flow directly to the next iteration
try:
    a = int(input("Enter a number:"))
    for i in range(11):
        if i == 0:
            print("Skip the iteration")
            continue # --> continue will skip rem code under current iteration 
        print(f"{a} X {i} = {a*i}")
except:
    print("Some error occured")
````

# Modules, packages, and libraries in python

1. A module is just a single .py file that contains pre-written code to perform specific tasks, and we can import 
and use these predefined functions and classes in our own code written by others to save time and effort.
  
There are 2 types of modules:  
- Built-in modules (os, turtle, random): Some modules come preinstalled in python. So u dont have to install them externally through pip (package installer) command. You just have to import them in your code.
- External Modules (matplotlib.pyplot, pandas, sklearn.metrics): For some use-cases, you have to download modules externally via pip command. Its syntax is: "pip install [package name]".  

2. A package (sklearn, pandas, matplotlib) is a physical directory (folder) containing multiple related .py files (submodules).  

A submodule (matplotlib.pyplot, sklearn.metrics) is simply a module (single .py file) inside a package.  

3. A library (Pandas, Matplotlib, Scikit-learn) is a large collection of code built for a purpose. It usually contains many packages and modules.  

For example:  
Pandas --> for Data Analysis  
Matplotlib --> for Data Visualization  
Scikit-learn --> for Machine Learning  
Tensorflow --> for Deep Learning

Importing built-in modules (math, random):

````python
# There are different ways to import a module/submodule:

# 1) Basic import (import whole module) with aliasing

import math as m # --> we create an alias of math through 'as m', so we dont have to write the whole word 'math' in order to use it, instead we can write 'm'.
print(m.sqrt(3)) # --> to use a function or class of a module, u have to specify name of module followed by a '.' then func or class name.
````

````python
import random as r
print(r.randint(1,100))
````

````python
# 2) To import a specific function/class from a module: from [module name] import [function/class]

from math import sqrt
print(sqrt(16))
````

````python
from random import randint
print(randint(1,100))
````

````python
# 3) Wildcard import

from math import * # Here, '*' means all. So now we have imported all func/classes of 'math', instead of importing whole module.
print(sqrt(5))
````

Note: Its generally advised not to make habit of wildcard import, as sometimes names of classses/functions in a modules may collide  
with python built-in functions (e.g, min and max).

````python
from random import *
print(random()) # random number between 0.0 to 1.0
print(randint(1, 100))
````

In order to import a user-defined function or class from a .py file (module), we follow this syntax:  
"from [folder name (package)].[file name (submodule)] import [class/func name] as [alias name]"  

If .py file (module) in same folder:  
from [file name (module)] import [func/class name]

````python
# Importing user-defined func from a .py file (module) named 'test'
import test
print(test.sum(a=12,b=7))
````

One useful point to note:  

'pandas' is a package, not a module or a submodule.  
So why dont we:  
import pandas.[submodule name]  
instead of:  
import pandas  

That is because pandas is designed so that its most important classes and functions are exposed at the top level.  

A useful rule:  
- If functionality is exposed by the package itself → import package
- If functionality lives in a specific submodule → import package.submodule
