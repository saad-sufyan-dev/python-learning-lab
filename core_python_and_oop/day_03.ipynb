# Introduction to File I/O in Python

I/O --> Input/Output, reading and writing data to files.

Volatile v/s Non-volatile Storage:
1. RAM --> Volatile(temporary), which is where active runtime data, program execution states and variables of a program are stored,  
    but data is lost when the power is turned off.
2. Hard Disk/SSD --> Non-volatile, which retains data even when the power is turned off.

We generally use files to store data on the hard disk/SSD, which allows us to save data permanently and access it later.

```python
# To open a file using python, we can use the built-in `open()` function.
f = open("demo.txt", "r")  # "r" stands for read mode
content = f.read()  # Read the entire content of the file
print(content)  # Print the content of the file

f.close()  # Close the file after reading
```

```python
# Using readline() to read a file line by line
b = open("demo.txt")
line1 = b.readline()
line2 = b.readline()
print(f"This is line 1 of demo.txt: {line1}") # Space will come after line one because of hidden \n (used for next line)
print(f"This is line 2 of demo.txt: {line2}")
b.close()
```

```python
# Using "w" mode (overwrite) and "a" mode (appending)

f = open("demo.txt","w")
f.write("I will learn file handling in python tomorrow.")
f.close()

b = open("demo.txt","a")
b.write("\nThen I'll upload it to Github.")
b.close()
```

```python
# If u open a file that doesnt exist using w or a mode, python automatically creates it for u
f = open("sample.txt","a")
f.write("This file was created automatically using 'a' mode")
f.close()
```

Meaning of different combinations of modes:  
"r" --> read mode (default), stream is positioned at the beginning of the file, throws error if file doesn't exist  
"w" --> write mode, stream is positioned at the beginning of the file and truncates the file to zero length, creates file if it doesn't exist  
"a" --> append mode, stream is positioned at the end of the file, creates file if it doesn't exist  
"r+" --> read and write mode, stream is positioned at the beginning of the file, throws error if file doesn't exist  
"w+" --> write and read mode, stream is positioned at the beginning of the file and truncates the file to zero length, creates 
file if it doesn't exist  
"a+" --> append and read mode, stream is positioned at the end of the file, creates file if it doesn't exist.


```python
# Using w+ mode
f = open("demo.txt","w+")
print(f.read()) # return nothing as file is truncated
f.write("Python is an amazing language.")
f.close()

# Using a+ mode
f = open("demo.txt","a+")
f.write("\nBecause of its readability.")
print(f.read()) # return nothing as pointer is placed at the end of file
f.close()

# Using r+ mode
f = open("demo.txt", "r+")
f.write("It also has a vast amount of libraries to help u solve problems")
print(f.read()) # return nothing as pointer is placed at the end of file after overwriting, need to use f.seek(0) to move pointer to the beginning.
f.close()
```

 When we use `with` statement to open a file, it automatically takes care of closing the file after the block of code is executed.


```python
# Using `with` statement to open a file
with open("demo.txt","r") as f:
    content = f.read()
    print(content)
```

```python
# Using os module to delete a file
import os
os.remove("sample.txt")
```

```python
# Connecting File I/O with OOP.
f = open("demo.txt","r") # creates a file obj
print(f.read()) # .read() method defined in class _io.TextIOWrapper
print(type(f))
f.close()
```

Let's dive deeper 🌟 and explore some key data structures: lists, tuples, dicts, and sets!
- List is a collection of items which is ordered (indexing) and mutable (changeable), and allows duplicate members.  
 It is defined using square brackets [].
- Tuples are similar to lists, but they are immutable (unchangeable) and defined using parentheses ().


```python
a = ["apple", "banana", "cherry", "apple"]
print(a)
print(a[2])
a[3] = "orange" # changing value at index 3
print(a)
print(a[:3]) # slicing a list

b = ("apple", "banana", "cherry", "apple")
print(b)
print(b[2])
# b[3] = "orange" # this will throw an error as tuples are immutable

print(type(a)) # <class 'list'>
print(type(b)) # <class 'tuple'>
```

List comprehensions are one of the most-loved Python language features.  
They enable you to combine for loop and if statement in one single line,  
which allows you to concisely form a new list by filtering the elements of a collection, transforming
the elements passing the filter in one concise expression. They take the basic form:  
[expr for val in collection if condition]

```python
# Suppose you wanted to filter out strings from a list of length 2 or less and also uppercase them in one statement
strings = ["a","as","bat","doodle"]

# So instead of using a for loop like this:
result = []
for i in strings:
    if len(i)<=2:
        result.append(i.upper())
print(result)

# We apply list comprehension which handles this task in a single statement
[x.upper() for x in strings if len(x)<=2]
```

```python
# Using a for loop through list comprehension
zeroes = [0 for x in range(10)] # creates a list of 10 zeroes
print(zeroes)
```

```python
# Concatenating strings
sentences = ["I like python", "Python is great", "I will learn python"]
print( ". ".join(sentences))
# Sorting lists
a = [4,2.4,5,1,0]
a.sort()
print(a)
```

```python
# Using del() and pop() methods
p = [2,3,5,7,11,13,17,18,23]
c = [4,6,8]

a = [1,2,"hi", True, p[3:6]]
b = ("elo", True, c)
del(a[0])
# del(b[1]) --> tuple's items cant be deleted
print(a, b)
a.pop(1) # returns item at given index and deletes it from list 
p[7] = 19
c.append(9)
print(c, b) # If tuple contains a mutable data typle (like a list), it will be updated without an error
```

- Dictionaries are a collection of key-value pairs which is ordered and mutable (changeable), and does not allow duplicate members. 
Its defined using curly braces {}.
- Sets is the collection of unique items which is unordered and immutable, and does not allow duplicate members. It is defined using curly braces {}.


```python
c = {"name": "Alice", "age": 25, "city": "New York", "is_adult": True, "children": ["Bob", "Charlie"]} # --> 'name' is a key and its value is 'Alice'.
print(c)
print(c["name"])
c["age"] = 26 # changing value of key "age"
print(c)
```

In dictionaries, we can have values of any data type, including lists and even other dictionaries. But keys must be of immutable data type,  
which means that we can use strings, integers, floats and tuples as keys, but not lists or dictionaries.  
  
Nested Dictionary: 
We can also convert our value to a dictionary

```python
info = {
    "name": "Saad Sufyan",
    "subjects": {
        "physics": 89,
        "chem": 95,
        "math": 100
    }
}

print(info["subjects"])
print(info["subjects"]["physics"]) # to access value of key "physics" inside the nested dictionary "subjects"

null_dict = {} # empty dictionary
print(null_dict)
print(type(null_dict))
```

```python
# Creating a dictionary for storing Sales Data
data = {"Item": ["Earbuds, Headphones", "Phone Charger"], "Price": [900, 1250, 245]} # keys act as column names whereas values represent rows (stored as a list)
print(data)

def categorize(price):
    result = []
    for i in price:
        if i>1000:
            result.append('A')
        elif i > 500:
            result.append('B')
        else:
            result.append('C')

    return result

data["Rating"] = categorize(data["Price"])
print(data)
```

```python
collection = {"apple", "banana", "cherry", "apple", 1, 2, 3, 1}
print(collection) # it will only print unique values, and order is not guaranteed

null_set = set() # empty set
print(null_set)
print(type(null_set))
```

```python
# Sets are often used to find out distinct items in a list
grocery_list = ["dog_food", "milk", "veggies", "oil"]
for item in grocery_list:
    print(f"letters used in {item} are {set(list(item))}")
```

```python
# Sets support mathematical set operations like union, intersection, difference, and symmetric difference.
a = {1,2,3,4}
b = {4,5,6}
print(a.union(b))
print(a.intersection(b))
print(f"{a.difference(b)} U {b.difference(a)} = {a.symmetric_difference(b)} ")
```

Immutable data types: string, integer, float, boolean, tuple --> value can be overwritten but not modified<br>
Mutable data types: list, dictionary, set

### Sets vs. Dictionaries

* **Both are mutable**: You can add, remove, or modify elements in a set and key-value pairs in a dictionary after creation.
* **Both rely on uniqueness**: Sets do not allow duplicate values, and dictionaries do not allow duplicate keys.
* **Both use curly braces**: Both structures are initialized using `{}` syntax (though an empty `{` defaults to a dictionary).

---

### Lists vs. Tuples

* **Both keep order**: Both structures preserve the exact sequence in which items are inserted.
* **Both allow duplicates**: You can store the same value multiple times at different positions in both structures.
* **Both support index access**: Elements in both collections are retrieved using the exact same zero-based indexing syntax, like `data[0]`.  

---

### Sets vs. Dictionaries (Differences)

* **Data Structure**: A set stores single, independent values (`{1, 2, 3}`), whereas a dictionary stores data as key-value pairs (`{'a': 1, 'b': 2}`).
* **Empty Initialization**: An empty dictionary is created using empty curly braces (`{}`), while an empty set must be initialized using the `set()` function.
* **Primary Utility**: Sets are designed for mathematical operations (like unions and intersections), whereas dictionaries are designed for fast data retrieval using mapped keys.

---

### Lists vs. Tuples (Differences)

* **Mutability**: A list is mutable, meaning you can add, remove, or change items. A tuple is completely immutable; its content cannot be altered after creation.
* **Syntax**: Lists are defined using square brackets `[]`, while tuples use parentheses `()`.
* **Memory Overhead**: Lists take up more memory because Python over-allocates extra space to handle dynamic resizing. Tuples allocate the exact amount of memory needed, making them more lightweight.
