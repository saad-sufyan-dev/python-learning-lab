# Introduction to Data Analysis with Pandas

Pandas is an open-source Python library built on top of NumPy, first released by Wes McKinney in 2008.
It is a high-level library known for its efficiency in handling structured data. Pandas supports multiple data workflows, including data cleaning, transformation, and analysis, which is why it is essential for data science and artificial intelligence.
It has a large ecosystem of integrations, making it suitable for a wide range of applications, from financial modeling and business analytics to preprocessing features for machine learning models.

Originally, there are 2 core data structures in Pandas:

- Series (1-Dimensional):

The data is structured as a single column of values, capable of holding any data type (integers, strings, floats, etc.).

Every element is explicitly mapped to an axis label, known as the index.

Mathematical operations are vectorized, allowing you to perform calculations across the entire series instantly without writing loops.

- DataFrame (2-Dimensional):

The entire data collection is structured as a tabular spreadsheet with distinct rows and columns.

It is exceptionally faster than native Python collections because the underlying data is stored in contiguous memory blocks.

However, managing memory can become a challenge with massive datasets, as a simple operation can create unintended copies of the entire structure in memory.

```python
# Importing libraries
import pandas as pd
import numpy as np
from numpy import nan as Na
```

<h3> Working with Series

```python
# Creating a series
s = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
print(s.index) # an attr returning index of series

data = pd.Series([[1,2,3],[4,5,6],2,5,{"name": "Huzaifa", "age":27}])
data
```

Think of a Series as a column in Excel. When you do: s*2, Pandas automatically multiplies every cell in that column by 2.  
This behavior is called **vectorized operations**.

```python
# Performing mathematical operations using series
s = s**3 # cube of every item
s
```

```python
# Selecting or modifying an element
s['a'] = 0
s
```

```python
print(f"Sum: {s.sum()}") # aggregate func which calculates sum of all items
print(f"Avg: {s.mean()}")
```

```python
# Operations between 2 series
s1 = pd.Series([1,2,3])
s2 = pd.Series([10,20,30, 0])
s3 = s1 + s2 # adds s1 element to its corresponding s2 element (s1[0]+s2[0] = s3[0])
s3
```

<h3> Dealing with DataFrames

```python
# Creating a DataFrame
df = pd.DataFrame({
    "name": ["Ali", "Sarah", "George"],
    "marks": [76,93,56]
})
df
```

```python
# Creating new columns named "grade" and "status" using map function and replace
def grade(m):
    if m >=90:
        return "A"
    elif m >=80:
        return "B"
    elif m>=70:
        return "C"
    elif m>=60:
        return "D"
    else:
        return "U (Ungraded)"
    
df["grade"] = list(map(grade, df["marks"]))
df["Status"] = list(map(lambda x: x>=60, df["marks"]))
df["Status"] = df["Status"].replace({True: "Promoted", False : "Failed"})
df
```

```python
# Using map for data transformation
data = pd.DataFrame({'food': ['bacon', 'pulled pork', 'bacon',
 'Pastrami', 'corned beef', 'Bacon',
 'pastrami', 'honey ham', 'nova lox'],
 'ounces': [4, 3, 12, 6, 7.5, 8, 3, 5, 6]})
 
data["food"] = data["food"].str.lower()

meat_to_animal = {
'bacon': 'pig',
'pulled pork': 'pig',
'pastrami': 'cow',
'corned beef': 'cow',
'honey ham': 'pig',
'nova lox': 'salmon'
}

data["meat_to_animal"] = data["food"].map(lambda x: meat_to_animal[x]) 
data
```

```python
# Concatenating 2 datasets
df2 = pd.DataFrame({
    "name": ["Daniel", "Mustufa"],
    "marks": [34,98],
})
df2["grade"]= list(map(grade, df2["marks"]))

final_df = pd.concat([df,df2])
final_df
```

```python
# Merging datasets
df = pd.DataFrame({
    "ID": [1000,1001,1002],
    "name": ["Ali", "Sarah", "Affan"]
})
df2 = pd.DataFrame({
    "name": ["Ali", "Sarah", "Affan"],
    "marks": [78,98,65]
})
final_df = pd.merge(df,df2,on="name")
final_df
```

```python
df = pd.DataFrame(np.random.randn(7,3)) # generates a dataframe (values normally ditributed) with 7 rows and 3 columns
df
```

```python
df.iloc[:3,1] = Na # .iloc is used to locate or select a range of rows (from 0 to 2) in a specific column (1)
df.iloc[:4, 2] = Na
df.iloc[:1, 0] = Na
df
```

```python
df.dropna() # it removes any row that contains null values
```

```python
df.dropna(thresh = 2) # now it only deletes rows which contain null values greater than or equal to 2
```
