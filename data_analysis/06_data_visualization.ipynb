# Basic Data Visualization with Pandas, Matplotlib, and Seaborn

Data visualization is the graphic representation of information and data. By using visual elements like charts, graphs, and maps, visualization tools provide an accessible way to see and understand trends, outliers, and patterns in data.

In the Python data science ecosystem, visual data analysis relies on a layered stack of libraries:

* **Matplotlib** serves as the foundational low-level plotting engine.
* **Pandas** provides high-level plotting convenience built directly on top of DataFrame and Series structures.
* **Seaborn** acts as a high-level statistical visualization layer built on Matplotlib, offering refined default styles and complex dataset-oriented plots.

---

### Matplotlib

Matplotlib provides complete control over every graphical element of a plot, including axes, ticks, labels, legends, and figure dimensions. It operates using two primary concepts:

* **`Figure`**: The overall window or page where everything is drawn. It can contain one or multiple axes.
* **`Axes`**: The actual plot or subplot where data is rendered. It contains the $x$-axis, $y$-axis, tick marks, lines, shapes, and labels.

### Pandas Plotting

Pandas wraps Matplotlib's functionality directly into DataFrames and Series via the `.plot` accessor (such as `df.plot()` or `df.plot.scatter()`). It handles data alignment, column labeling, and index rendering automatically, making it ideal for fast, exploratory data analysis.

### Seaborn

Seaborn integrates tightly with Pandas data structures and automates complex statistical visualizations. It manages aesthetic themes, color palettes, and statistical transformations (such as computing confidence intervals or aggregation) internally, while utilizing Matplotlib to draw the output.

## Core Visualization Types and Applications

Selecting the appropriate plot depends on the data types (categorical vs. numerical) and the analytical question being addressed.

| Chart Type       | Target Data Types                         | Primary Purpose                                                             | Common Function Names                    |
| ---------------- | ----------------------------------------- | --------------------------------------------------------------------------- | ---------------------------------------- |
| **Line Chart**   | Numerical vs. Time/Ordered Numerical      | Shows trends over continuous time or ordered intervals                      | `plot()`, `lineplot()`                   |
| **Bar Chart**    | Categorical vs. Numerical                 | Compares discrete categories by aggregate values                            | `plot.bar()`, `barplot()`, `countplot()` |
| **Histogram**    | Continuous Numerical                      | Displays the underlying frequency distribution of a single variable         | `plot.hist()`, `histplot()`              |
| **Scatter Plot** | Numerical vs. Numerical                   | Reveals relationships, correlations, and clusters between two variables     | `plot.scatter()`, `scatterplot()`        |
| **Box Plot**     | Categorical vs. Continuous Numerical      | Summarizes distribution using five-number summaries and highlights outliers | `plot.box()`, `boxplot()`                |
| **Heatmap**      | 2D Numerical Matrix or Correlation Matrix | Visualizes data intensity using color gradients                             | `heatmap()`                              |
| **Pie Chart**    | Categorical + Numerical                   | Shows how each category contributes to a whole                              | `plot.pie()`, `pie()`                    |

## Detailed Chart Analysis

### 1. Line Charts (Trend Analysis)

Line charts connect individual data points with line segments. They are best suited for continuous variables, particularly time series analysis.

* **Mechanism**: Maps independent variables (often time) to the horizontal axis and dependent metrics to the vertical axis.

* **Key Considerations**: Avoid plotting too many overlapping lines, as this creates visual clutter known as the **"spaghetti plot" effect**.

### 2. Bar Charts and Count Plots (Categorical Comparison)

Bar charts use rectangular bars where the length of each bar is proportional to the values represented.

* **Bar Plot**: Computes a summary statistic (like mean or sum) for a numerical variable across different categories.

* **Count Plot**: Counts how many observations belong to each category.

* **Key Considerations**: Always start the vertical axis at zero to prevent visually exaggerating differences between categories.

### 3. Histograms and KDE Plots (Distribution Analysis)

Histograms divide continuous numeric data into consecutive, non-overlapping intervals called **bins**, counting how many values fall into each bin.

* **Binning Strategy**: The number or width of bins changes the appearance of the distribution. Too few bins oversimplify the data; too many bins create excessive noise.

* **Kernel Density Estimation (KDE)**: A smooth curve that shows the overall shape of a numerical distribution.

### 4. Scatter Plots (Relationship and Correlation)

Scatter plots display individual values using dots positioned on Cartesian axes to represent two numerical variables simultaneously.

* **Mechanism**: Identifies linear or non-linear patterns, strength of association, clusters, and unusual values (outliers).

* **Overplotting**: When too many data points overlap, the distribution of points becomes difficult to see. Solutions include reducing point opacity (`alpha`), reducing point size, or using hexagonal binning.

### 5. Box Plots (Summary Statistics and Outlier Detection)

Box plots (or box-and-whisker plots) visually show the distribution of quantitative data through five key statistics:

1. **Minimum**: Lowest non-outlier value within the lower threshold.
2. **First Quartile (Q1)**: 25th percentile of the data.
3. **Median (Q2)**: 50th percentile of the data.
4. **Third Quartile (Q3)**: 75th percentile of the data.
5. **Maximum**: Highest non-outlier value within the upper threshold.

The distance between Q3 and Q1 is the **Interquartile Range (IQR)**:

\(IQR = Q3 - Q1\)

Data points extending beyond `1.5 × IQR` from either quartile are typically plotted as individual points representing potential outliers.

### 6. Heatmaps (Multivariate Matrices)

Heatmaps represent values within a matrix using colors. They are frequently used to inspect correlation matrices (`df.corr()`) across multiple numerical features.

* **Color Mapping**: Uses sequential colors for values progressing from low to high, or diverging colors when showing values on both sides of a meaningful center point such as zero.

### 7. Pie Charts (Part-to-Whole Analysis)

Pie charts divide a circle into slices, where each slice represents a category's **share of the total**.

* **Mechanism**: The size of each slice is proportional to that category's contribution to the whole.

* **Best Use**: Shows simple proportions when there are only a few categories.

* **Example**: Showing how much each product category contributes to total sales.

* **Key Consideration**: Avoid using pie charts with too many categories because comparing many slices becomes difficult. For precise comparisons, a bar chart is usually better.
---

## Library Selection Guidelines

Choosing which library to use depends on the analytical phase and output requirements.

* Use **Pandas Plotting** when conducting initial data exploration directly on DataFrames, prioritizing speed over visual customization.
* Use **Seaborn** when analyzing statistical relationships, exploring categorical interactions, or building polished visuals for presentations.
* Use **Matplotlib** directly when building multi-panel figure layouts, creating non-standard visual structures, or explicitly modifying minor details like custom axes and annotations.

---

## Common Mistakes to Avoid

* **Truncating the Y-Axis**: Starting vertical axes at non-zero values on bar charts, which visually exaggerates minor relative differences.
* **Ignoring Overplotting**: Plotting thousands of dense scatter points without transparency (`alpha`), making density assessment impossible.
* **Selecting Poor Color Palettes**: Using rainbow or non-uniform color schemes that misrepresent continuous numerical progressions or obscure patterns for color-blind viewers.
* **Misinterpreting Correlation as Causation**: Assuming a visible pattern or linear trend in a scatter plot implies a direct causal relationship between the variables.

````python
# Plotting sales line chart

fig, ax = plt.subplots(1, 1, figsize=(12, 5))

sales_by_year = df.groupby("Transaction Year")["Total Spent"].sum()
ax.plot(sales_by_year, marker="o", linewidth=2, color="green")
ax.set_title("Sales Trend by Year")
ax.set_xlabel("Year")
ax.set_xticks(df["Transaction Year"])
ax.set_ylabel("Sales")
ax.grid(True, alpha=0.3)


plt.tight_layout()
plt.show()
````
