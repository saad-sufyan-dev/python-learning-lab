# Basic Data Visualization

## Definition

**Data visualization** is the graphic representation of information and data. By using visual elements like charts, graphs, and maps, visualization tools provide an accessible way to see and understand trends, outliers, and patterns in data.

In Python, data visualization relies on a tiered ecosystem:

* **`matplotlib.pyplot`**: The foundational, low-level plotting engine that offers explicit control over figure elements, axes, tick marks, and layout mechanics.


* **`pandas`**: Offers high-level, convenient plotting wrappers directly on DataFrames and Series, designed for quick data exploration.


* **`seaborn`**: A high-level visualization library built on top of Matplotlib that integrates seamlessly with Pandas DataFrames to generate statistically oriented, publication-ready graphics with polished default styles.



---

## Why It Matters

Raw numerical tables and descriptive statistics can obscure critical patterns, relationships, or anomalies. Visualizing data helps analysts:

* **Identify Patterns and Trends**: Quickly spot linear or non-linear relationships, seasonality, and clusters.


* **Detect Outliers and Anomalies**: Instantly recognize severe data-entry errors or extreme skewness that might distort summary metrics.


* **Validate Analytical Assumptions**: Check distributions for normality or homoscedasticity before applying statistical models or machine learning algorithms.


* **Communicate Insights Effectively**: Present complex analytical findings to non-technical stakeholders in a clear, intuitive manner.

## Selecting the Right Chart Type

Choosing an appropriate plot depends on the data types (numerical vs. categorical) and the analytical question being addressed:

* **Distribution of a Single Variable**
* **Histogram**: Displays the frequency distribution of a continuous numerical variable by grouping data into bins.


* **KDE (Kernel Density Estimate) Plot**: Displays a smooth, continuous probability density curve for a numerical variable.


* **Box Plot**: Shows distribution summaries using quartiles, medians, and explicit outlier points.




* **Relationships Between Variables**
* **Scatter Plot**: Plots individual data points along two continuous axes to inspect relationships, correlation, or clustering.


* **Line Plot**: Connects data points sequentially, making it ideal for time-series data and trend analysis.




* **Categorical Comparisons**
* **Bar Chart**: Compares aggregated numerical metrics across distinct categories.


* **Count Plot**: Displays the frequency of occurrences for discrete categorical groups.

## Visualization Workflow

A systematic visualization workflow combines high-level plotting with explicit canvas manipulation:

1. **Initialize Canvas and Axes**
* Define explicit figure dimensions using `plt.subplots()` to ensure full structural control over layout and grid setups.




2. **Generate Plot via Library Wrappers**
* Call `seaborn` statistical plotting functions or `pandas` DataFrame plotting methods, passing the target `ax` (axes) object explicitly.




3. **Customize Visual Hierarchy**
* Apply explicit labels, titles, tick formatting, legends, and grid lines using `matplotlib.pyplot` methods.




4. **Adjust Layout and Render**
* Use layout adjustments like `plt.tight_layout()` to prevent label clipping before rendering or saving the figure.

## Common Mistakes

* **Creating Unlabelled or Ambiguous Axis Labels**: Charts without clear units, legends, or descriptive axis titles lead to misinterpretation.


* **Truncating Axes to Distort Data**: Forcing axis limits away from baseline zeros in bar charts can artificially exaggerate minor differences between categories.


* **Overcrowding Figures**: Plotting too many categories, markers, or trendlines simultaneously creates visual noise.


* **Selecting Inappropriate Chart Types**: Representing continuous time-series data with unsorted bar charts or categorical frequencies with scatter plots hides logical structures.


* **Mixing Plotting Styles Without Purpose**: Combining arbitrary colors, themes, or uncoordinated figure dimensions distorts visual communication.

## Quick Recap

* **`matplotlib`** provides the core canvas and granular structural control.


* **`pandas`** offers quick, native plotting directly from Series and DataFrames.


* **`seaborn`** simplifies complex, publication-ready statistical visualizations.


* **Chart selection** must strictly reflect the underlying variable types and analytical objectives.


* **Always label axes, legends, and titles** to maintain clarity and technical accuracy.
