# Basic Data Cleaning

## Definition

**Data cleaning** is the process of **detecting, correcting, removing, or handling** inaccurate, incomplete, inconsistent, duplicate, or incorrectly formatted data in a dataset.

In Python, **Pandas** is one of the primary libraries used to inspect, transform, and clean tabular data.

---

## Why It Matters

Real-world data is rarely ready for analysis. It may contain:

- Missing values
- Duplicate records
- Incorrect data types
- Inconsistent categories or formatting
- Invalid values
- Data-entry errors
- Outliers
- Unnecessary or irrelevant data

If these issues are not handled appropriately, they can lead to **incorrect calculations, misleading insights, and unreliable models**.

Therefore, data cleaning is an essential step before performing reliable analysis or building machine learning models.

---

## Data Cleaning Workflow

A typical data-cleaning process follows a systematic workflow:

1. **Inspect the Data**
   - Examine shape, columns, data types, and sample records.
   - Identify missing values, duplicates, and suspicious values.

2. **Handle Missing Values**
   - Determine why values are missing.
   - Depending on the situation, remove them or replace them using appropriate values such as the mean, median, mode, or a business-specific value.

3. **Handle Duplicate Records**
   - Identify repeated records.
   - Remove duplicates when they represent unintended repetition.

4. **Correct Data Types**
   - Ensure each column has an appropriate data type.
   - For example, convert text-based dates to `datetime` and numeric strings to numeric values.

5. **Standardize Inconsistent Data**
   - Fix inconsistent capitalization, whitespace, spelling, units, or categorical labels.
   - Example: `"Male"`, `"male"`, and `" M "` may need to be standardized.

6. **Handle Invalid Values**
   - Identify values that violate logical or business rules.
   - Example: a retail dataset containing `-5` units sold.

7. **Validate the Cleaned Data**
   - Recheck missing values, duplicates, data types, ranges, and business rules.
   - Make sure the cleaning process did not introduce new problems.

---

## Common Data Cleaning Tasks

| Cleaning Task | Purpose | Common Pandas Approach |
|---|---|---|
| **Missing Values** | Identify and handle incomplete records | `isna()`, `dropna()`, `fillna()` |
| **Duplicate Rows** | Remove unintended repeated records | `duplicated()`, `drop_duplicates()` |
| **Data Type Conversion** | Ensure columns have appropriate types | `astype()`, `pd.to_numeric()`, `pd.to_datetime()` |
| **Text Standardization** | Make text values consistent | `.str.strip()`, `.str.lower()`, `.str.upper()` |
| **Invalid Values** | Detect values that violate rules | Boolean filtering, conditions |
| **Outliers** | Identify unusually high or low observations | Statistical methods, domain rules |
| **Unnecessary Data** | Remove irrelevant columns or records | `drop()`, filtering |

---
## Common Mistakes in Data Cleaning

* **Modifying the raw dataset directly**
  * Always preserve the original data so you can reproduce or undo your cleaning process.

* **Dropping missing values blindly**
  * Removing rows can result in significant information loss.

* **Filling missing values without considering context**
  * Mean, median, mode, or other methods are not universally appropriate.

* **Ignoring data types**
  * Numbers stored as strings cannot be reliably used for numerical analysis.

* **Removing duplicates without understanding them**
  * Some repeated-looking records may represent legitimate transactions.

* **Deleting outliers automatically**
  * An outlier may represent a genuine and important observation rather than an error.

* **Cleaning without validation**
  * A dataset should be checked again after cleaning to confirm that the intended problems were actually resolved.

---

### `.any()` and `.all()` in pandas

#### Definition

In Pandas, `.any()` and `.all()` evaluate Boolean conditions across a Series or DataFrame:

* **`.any()`**: Returns `True` if **at least one** value in the evaluated axis is `True` (logical **OR**).


* **`.all()`**: Returns `True` only if **every** value in the evaluated axis is `True` (logical **AND**).



---

#### How It Works

Both methods evaluate Boolean flags along a target axis:

* **`axis=0` (Default)**: Evaluates vertically down each column.


* **`axis=1`**: Evaluates horizontally across each row.



To optimize execution, both use short-circuit evaluation: `.any()` stops at the first `True`, while `.all()` stops at the first `False`.

---

#### Common Mistakes

* **Forgetting `axis=1`**: Running checks across rows without specifying `axis=1` evaluates columns instead.


* **Ignoring Empty Objects**: `.all()` on an empty Series returns `True` because no `False` value exists to fail the check.
* **Confusing Numerical Data**: Non-zero numbers evaluate to `True` and `0` evaluates to `False`, which can produce unintended results if non-Boolean data is evaluated directly.

---

#### Quick Recap

* **`.any()`** checks if **at least one** condition is `True` (useful for finding anomalies or missing data).


* **`.all()`** checks if **every** condition is `True` (useful for validating full compliance).


* Both methods evaluate **columns (`axis=0`)** by default.

````python
# Inspecting the null-affected rows
df[df.isna().any(axis=1)]
````

````python
# Filling related numerical columns
df["Price Per Unit"] = df["Price Per Unit"].fillna(
    df["Total Spent"] / df["Quantity"]
)

df["Quantity"] = df["Quantity"].fillna(
    df["Total Spent"] / df["Price Per Unit"]
)

df["Total Spent"] = df["Total Spent"].fillna(
    df["Price Per Unit"] * df["Quantity"]
)
````

### The `subset` parameter in `dropna()` and `drop_duplicates()`

The ```subset``` parameter in ```dropna()``` and ``drop_duplicates()`` tells Pandas:

"Only look at these specific columns when deciding what to remove."  

So if `dropna(subset=['Transaction ID'])`, Pandas only checks the `'Transaction ID'` column for nulls, while other columns are ignored

````python
# Dropping the nulls which we cant derive using subset and thresh parameter
df = df.dropna(
    subset=["Price Per Unit", "Quantity", "Total Spent"],
    thresh=2
)
````

````python
# Derive missing Item values using the relationship discovered earlier

# Create a lookup table mapping Category + Price Per Unit → Item
item_map = (
    df.dropna(subset=["Item"])
      .drop_duplicates(subset=["Category", "Price Per Unit"])
      .set_index(["Category", "Price Per Unit"])["Item"]
)

# Identify rows where Item is missing
mask = df["Item"].isna()

# Fill missing Item values using Category + Price Per Unit
df.loc[mask, "Item"] = [
    item_map.get((category, price))
    for category, price in zip(
        df.loc[mask, "Category"],
        df.loc[mask, "Price Per Unit"]
    )
]
````

### Outlier Detection: IQR Method vs. 3 Sigma Rule



Outliers are extreme values that deviate significantly from the rest of the dataset. Detecting them prevents skewed calculations and model distortion.

---

### Interquartile Range (IQR) Method



The IQR method uses percentiles to identify outliers. It works well on skewed distributions because it relies on median-based statistics.

* **IQR** $= Q3 - Q1$

* **Lower Bound** $= Q1 - (1.5 \times \text{IQR})$

* **Upper Bound** $= Q3 + (1.5 \times \text{IQR})$


Data points falling outside these bounds are flagged as outliers.

---

### 3 Sigma Rule (Z-Score)



The 3 Sigma rule relies on the mean and standard deviation. It assumes the dataset follows a normal distribution.

* **Lower Bound** $= \mu - 3\sigma$

* **Upper Bound** $= \mu + 3\sigma$


Values beyond three standard deviations from the mean ($\approx 99.7\%$ of normal data) are flagged.

---

### Comparison



| Method | Assumption | Best Used For |
| --- | --- | --- |
| **IQR Method** | Non-parametric (no distribution assumption) | Skewed or non-normal data |
| **3 Sigma Rule** | Parametric (assumes normal distribution) | Symmetric, bell-curved data |

````python
# Outlier detection through IQR Method

Q1 = df["Total Spent"].quantile(0.25)
Q3 = df["Total Spent"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5*IQR)
upper_bound = Q3 + (1.5*IQR)

outliers = df[
    (df["Total Spent"]<lower_bound) |
    (df["Total Spent"]>upper_bound)
]
````
