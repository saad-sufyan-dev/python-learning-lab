# Exception Handling in python

Generally, when we execute a python program, we use a python interpreter (it translates code to binary line by line) as python is an interpreted  
language, unlike C and C++.  

When a python interpreter encounters an error while interpreting (such as converting a string into an integer), it returns to the program and  
checks if user has provided what to do if such error occured (how to encounter ValueError).  
If yes, it handles error accordingly, otherwise, it crashes and terminates flow of program.  

For Example, if we have 10 lines of code and error happens at line 5, then flow of program will terminate at line 5.  
Thats why its imp to learn how to handle exceptions in python.

````python
# Using try and except 
try:
    a = int(input("Enter an integer: "))
    print(f"Multiplication Table of {a} is:")
    for i in range(1, 11):
        print(f"{a} X {i} = {a * i}")
except ValueError:
    print("Input not an integer!")
except:
    print("Some error occured.")
    
print("\nSome imp lines of code.")
# If we enter a number, code will execute normally. But if someone enters a string, instead of crashing the program the interpreter would 
# handle that exception (printing that input is invalid) and successfully execute whole program without halting in the middle.
````

````python
# In python we can also raise errors and halt the program execution, like when a situation breaks real-life logic,
# even if it doesnt trigger a standard system crash. 

# Using raise keyword

def withdraw_money(balance, amount):
    if amount <= 0:
        raise ValueError("Withdrawal amount must be greater than zero.")
    
    if amount > balance:
        # Raising an error because you cannot withdraw more than you own
        raise PermissionError("Insufficient funds for this transaction.")
    
    return balance - amount

try:
    print(withdraw_money(5000, 6000))

except PermissionError as error:
    print(f"Transaction Denied: {error}")

except:
    print("Some error occured.")
````

```text
[ Code Execution ]
 └── Call: withdraw_money(5000, 6000)
      ├── Condition Check: 6000 > 5000 (True)
      └── raise PermissionError ──> (Execution Halts)
                                           │
[ Exception Handling ]                     ▼
 └── caught by 'except PermissionError as error'
      └── Prints: "Transaction Denied: Insufficient funds for this transaction."

````python
a = input("Enter a number: ")
try: 
    a = int(a)
    b = 1/a
    print(f"1 divided by {a} is {b}")

except ZeroDivisionError: 
    print("Cant divide by 0!")

except ValueError:
    print("Invalid input, not an integer")

except Exception as e: # here storing any other error in an alias 'e' and handling it using except 
    print(f"Some error occured:\n{e}")
````

```text
Syntax for try-except block:

try:  
    # risky code  
except:  
    # handle error  
finally:  
    # always runs

````python
# the finally clause in try-except block: it always runs whether an error occurs or not
# Common use-case: Closing a file

file = None
try:
    file = open("data.txt")
    data = file.read()
except FileNotFoundError:
    print("File not found")
finally:
    if file != None:
        file.close()

# Usually we use 'with' syntax, which automatically closes file
try:
    with open("data.txt") as f:
        data = f.read()

except FileNotFoundError:
    print("File not Found")
````

# Statements (break, continue and pass) in python

````python
# break statement terminates the loop completely and transfers flow of execution to the next statement outside the loop
# Usually used with for-else statement, where the else block belongs to for loop instead of if statement/
numbers = [1,2,3,4]

for num in numbers:
    if num == 4:
        print("Found")
        break # --> If val is found, 'break' will terminate the loop, and the else block will be skipped

else: # --> If no break statement executes during looping, then else block will execute 
    print("Not Found")
````

````python
# continue statement skips remaining code inside current iteration inside the loop and transfers flow directly to the next iteration
try:
    a = int(input("Enter a number:"))
    for i in range(11):
        if i == 0:
            print("Skip the iteration")
            continue # --> continue will skip rem code under current iteration 
        print(f"{a} X {i} = {a*i}")
except:
    print("Some error occured")
````

# Modules in python

A module is like a code library that contains pre-written code to perform specific tasks, and we can import 
and use these predefined functions and classes   
in our own code written by others to save time and effort.
  
There are 2 types of modules:  
- Built-in modules: Some modules come preinstalled in python, including os, math, etc. So u dont have to install them externally through pip (package installer) command. 
- External Modules: For some use-cases, like data manipulation or machine learning, u have to download modules externally via pip command. Its syntax is
                    "pip install [module name]".

````python
# There are different ways to import a module:

# 1) Basic import (import whole module) with aliasing

import math as m # --> we create an alias of math through 'as m', so we dont have to write the whole word 'math' in order to use it, instead we can write 'm'.
print(m.sqrt(3)) # --> to use a function or class of a module, u have to specify name of module followed by a '.' then func or class name.
````

````python
# 2) To import a specific function/class from a module: from [module name] import [function/class]

from math import sqrt as s 
print(s(16))
````

A submodule is simply a module inside a package.  
Think package as a big folder and submodules as files in it.  

So if you import math, its a module (single .py file)  
and if you import sklearn, its a package (a big folder containing many submodules (single .py files) or other subpackages)

```text
sklearn (package)
└── metrics (module/subpackage)
    └── classification_report (function)

from sklearn.metrics import classification_report

Python reads this program as:  
Go to the sklearn package → open the metrics submodule → import classification_report.

````python
# 3) Wildcard import

from math import * # --> Here, '*' means all. So now we can use all public func/classes and variables from 'math' directly, instead of -->math.sqrt(9)--> dont need to mention module name.
print(sqrt(5))

# Note: Its generally not preferred to use wildcard import bcz if u import 2 modules through it and if a func/class exists with the same name in both modules,
#   it will be difficult to know which module's func is it.
````

In order to import a user-defined function or class from a file, we follow this syntax:  
"from [folder name].[file name] import [class/func name] as [alias name]"  
If file in same folder:  
from [file name] import [func/class name]

````python
# Importing user-defined func from a .py file 
from test import sum
print(sum(99,67))
````
